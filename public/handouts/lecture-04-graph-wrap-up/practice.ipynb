# 题目一：P1339 热浪

[Luogu P1339](https://www.luogu.com.cn/problem/P1339)

## 1.1 题目说明

给定一张无向带权图，共有 $N$ 个城镇、$M$ 条双向道路。每条道路连接两个城镇，并有一个非负长度。给定起点 $S$ 与终点 $T$，要求输出从 $S$ 到 $T$ 的最短路长度。

输入可以理解为：

- 第一行：$N, M, S, T$
- 接下来 $M$ 行：$u, v, w$，表示 $u$ 与 $v$ 之间有一条长度为 $w$ 的双向边
- 输出：从 $S$ 到 $T$ 的最短距离

这是一道标准的**非负权单源最短路**题。虽然只问 $T$ 一个点，但算法本质仍然是从 $S$ 出发不断确定其它点的最短距离。


---

## 1.2 思路分析

令 $dist[x]$ 表示当前已知的从 $S$ 到 $x$ 的最短距离上界。初始时：

$$
dist[S] = 0, \qquad dist[x] = +\infty \quad (x \ne S)
$$

Dijkstra 的核心动作是**松弛**。若当前已经知道一条到 $u$ 的路径，再经过边 $(u,v,w)$，就得到一条到 $v$ 的候选路径：

$$
dist[v] \leftarrow \min\{dist[v],\ dist[u] + w\}
$$

为什么可以贪心？因为所有边权都满足 $w \ge 0$。当小根堆弹出当前最小的 $(d,u)$，且 $d=dist[u]$ 时，不可能再通过其它尚未确定的点绕出一条更短路径到 $u$。如果存在这样的路径，它必须先到达某个尚未确定的点 $x$，而这个前缀距离已经不小于当前最小值 $d$，再加上非负边权也不会变小。

### 实现要点

| 要点 | 说明 |
|---|---|
| 邻接表 | 稀疏图更适合 `g[u].append((v, w))` |
| 小根堆 | Python 用 `heapq`，每次弹出当前距离最小的点 |
| 懒删除 | 堆里可能有旧距离，遇到 `d != dist[u]` 直接跳过 |
| 提前结束 | 如果只求 $T$，第一次弹出 $T$ 时答案已经确定 |

复杂度为：

$$
O((N+M)\log N)
$$

空间复杂度为 $O(N+M)$。


---

## 1.3 Python 参考代码

下面这一格是完整提交版：从标准输入读取，向标准输出写答案。


```python
import sys
import heapq


def dijkstra(n, graph, start, target):
    INF = 10**30
    dist = [INF] * (n + 1)
    dist[start] = 0
    heap = [(0, start)]

    while heap:
        d, u = heapq.heappop(heap)
        if d != dist[u]:
            continue
        if u == target:
            return d
        for v, w in graph[u]:
            nd = d + w
            if nd < dist[v]:
                dist[v] = nd
                heapq.heappush(heap, (nd, v))

    return dist[target]


def solve(data: str) -> str:
    it = iter(map(int, data.split()))
    n = next(it)
    m = next(it)
    start = next(it)
    target = next(it)

    graph = [[] for _ in range(n + 1)]
    for _ in range(m):
        u = next(it)
        v = next(it)
        w = next(it)
        graph[u].append((v, w))
        graph[v].append((u, w))

    return str(dijkstra(n, graph, start, target))


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    print(solve(sys.stdin.read()))
```

---

## 1.4 测试验证

用一个经典样例结构验证：从 $1$ 到 $6$ 的最短路是 $1\to3\to6$，长度为 $9+2=11$，而不是直接边 $14$。


```python
sample = """
6 9 1 6
1 2 7
1 3 9
1 6 14
2 3 10
2 4 15
3 4 11
3 6 2
4 5 6
5 6 9
"""

assert solve(sample) == "11"
print("P1339 测试通过：最短路长度 =", solve(sample))
```

---

## 1.5 易错点与变体引出

这道题最值得积累的是“适用条件”。Dijkstra 不是万能最短路模板，它依赖非负权边；如果有负边，贪心弹出的点未必已经最优。

常见错误包括：

1. **把无向边只加了一次**：本题道路是双向的，必须同时加入 $u\to v$ 与 $v\to u$。
2. **没有跳过堆中的旧状态**：同一个点可能被多次压入堆，必须用 `if d != dist[u]: continue` 保证处理的是最新距离。
3. **错误使用 BFS**：只有所有边权相同，BFS 才能保证最短路；一般正权图要用 Dijkstra。
4. **负权边误用 Dijkstra**：只要存在负边，就应该换 Bellman-Ford/SPFA 等方法。

### 变体 1.5.1：边权只有 $0$ 或 $1$ —— 0-1 BFS

若边权 $w\in\{0,1\}$，可以用双端队列维护“当前层”和“下一层”。走 $0$ 边时放队首，走 $1$ 边时放队尾，复杂度降为：

$$
O(N+M)
$$

**核心代码：**


```python
from collections import deque


def zero_one_bfs(n, graph, start):
    INF = 10**30
    dist = [INF] * (n + 1)
    dist[start] = 0
    dq = deque([start])

    while dq:
        u = dq.popleft()
        for v, w in graph[u]:
            nd = dist[u] + w
            if nd < dist[v]:
                dist[v] = nd
                if w == 0:
                    dq.appendleft(v)
                else:
                    dq.append(v)
    return dist
```

### 变体 1.5.2：存在负权边但无负环 —— Bellman-Ford

如果允许负权边，Dijkstra 的贪心性失效。Bellman-Ford 用“反复松弛所有边”解决：一条最短路在没有负环时最多经过 $N-1$ 条边，因此最多做 $N-1$ 轮松弛。

**核心代码：**


```python
def bellman_ford(n, edges, start):
    INF = 10**30
    dist = [INF] * (n + 1)
    dist[start] = 0

    for _ in range(n - 1):
        changed = False
        for u, v, w in edges:
            if dist[u] != INF and dist[u] + w < dist[v]:
                dist[v] = dist[u] + w
                changed = True
        if not changed:
            break
    return dist
```

---

## 1.6 本题小结

P1339 是最短路学习中的“分水岭”：从无权图的 BFS 进入一般带权图。课堂上应记住三件事：

| 情况 | 推荐算法 | 关键判断 |
|---|---|---|
| 无权图 / 等权图 | BFS | 每走一条边代价相同 |
| 边权为 $0/1$ | 0-1 BFS | 双端队列维护距离层次 |
| 非负权图 | Dijkstra | 小根堆 + 贪心确定最短距离 |
| 有负边无负环 | Bellman-Ford | 反复松弛，复杂度较高 |

本题的核心收获：**最短路算法首先看边权结构，再选算法。**

# 题目二：P3367 【模板】并查集

[Luogu P3367](https://www.luogu.com.cn/problem/P3367)

## 2.1 题目说明

给定 $N$ 个元素和 $M$ 个操作，需要维护若干集合。操作分两类：

- `1 x y`：合并 $x$ 与 $y$ 所在的集合；
- `2 x y`：查询 $x$ 与 $y$ 是否在同一个集合中，是则输出 `Y`，否则输出 `N`。

这道题不直接给图，但它等价于动态加入无向边，并查询两个点是否连通。它是并查集（Disjoint Set Union, DSU）的模板题。


---

## 2.2 思路分析

并查集用一棵树表示一个集合。每个集合选一个代表元，称为根。我们维护数组：

$$
fa[x] = x \text{ 的父亲节点}
$$

若 $fa[x]=x$，说明 $x$ 是所在集合的根。判断两个元素是否在同一集合，只需比较：

$$
find(x) = find(y)
$$

### 路径压缩

如果每次 `find` 都沿父亲指针一路向上，树可能很深。路径压缩会在查询根的同时，把路径上的点直接挂到根上：

$$
fa[x] \leftarrow find(fa[x])
$$

这样后续查询会更快。

### 按大小合并

合并两个集合时，把较小的树挂到较大的树上：

$$
size[root_{large}] \leftarrow size[root_{large}] + size[root_{small}]
$$

路径压缩 + 按大小合并后，单次操作的均摊复杂度接近常数，通常写作：

$$
O(\alpha(N))
$$

其中 $\alpha$ 是反 Ackermann 函数，在实际数据范围内可以视为很小的常数。


---

## 2.3 Python 参考代码

这一格是完整提交版。注意本题 $M$ 可很大，输出应先收集到列表，最后一次性 `join`。


```python
import sys


class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n + 1))
        self.size = [1] * (n + 1)

    def find(self, x):
        while self.parent[x] != x:
            self.parent[x] = self.parent[self.parent[x]]
            x = self.parent[x]
        return x

    def union(self, x, y):
        rx = self.find(x)
        ry = self.find(y)
        if rx == ry:
            return False
        if self.size[rx] < self.size[ry]:
            rx, ry = ry, rx
        self.parent[ry] = rx
        self.size[rx] += self.size[ry]
        return True

    def same(self, x, y):
        return self.find(x) == self.find(y)


def solve(data: str) -> str:
    it = iter(map(int, data.split()))
    n = next(it)
    m = next(it)
    uf = UnionFind(n)
    ans = []

    for _ in range(m):
        op = next(it)
        x = next(it)
        y = next(it)
        if op == 1:
            uf.union(x, y)
        else:
            ans.append("Y" if uf.same(x, y) else "N")

    return "\n".join(ans)


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    print(solve(sys.stdin.read()))
```

---

## 2.4 测试验证

下面的操作先合并 $(1,2)$ 与 $(3,4)$，第一次查询 $1$ 与 $3$ 不连通；再合并 $(2,3)$ 后，$1$ 与 $4$ 就连通了。


```python
sample = """
5 5
1 1 2
1 3 4
2 1 3
1 2 3
2 1 4
"""

assert solve(sample) == "N\nY"
print("P3367 测试通过：")
print(solve(sample))
```

---

## 2.5 易错点与变体引出

并查集最容易错在“根”和“父亲”的概念混淆：只有 `find(x)` 才是集合代表元，`parent[x]` 可能只是中间父亲。比较集合时必须比较根。

还要注意：

1. **不要用列表存集合成员再合并**：一次合并可能移动大量元素，会退化到 $O(NM)$。
2. **递归版 `find` 可能触发递归深度限制**：Python 中迭代写法更稳。
3. **查询操作不改变集合关系**：`op=2` 只判断，不合并。

### 变体 2.5.1：Kruskal 最小生成树

把边按权值从小到大排序，依次尝试加入。若一条边连接的两个端点已经在同一集合中，加入它会形成环；否则加入 MST。

**核心代码：**


```python
def kruskal(n, edges):
    """edges: [(u, v, w), ...]；返回 MST 权重，若不连通则返回 None。"""
    uf = UnionFind(n)
    total = 0
    chosen = 0

    for u, v, w in sorted(edges, key=lambda e: e[2]):
        if uf.union(u, v):
            total += w
            chosen += 1
            if chosen == n - 1:
                return total

    return None
```

### 变体 2.5.2：朋友圈 / 连通块大小

如果题目要求“某个元素所在集合有多大”，并查集只需额外维护根节点的 `size`。合并时更新新根大小，查询时返回：

$$
size[find(x)]
$$

**核心代码：**


```python
def component_size(uf, x):
    return uf.size[uf.find(x)]


uf_demo = UnionFind(5)
uf_demo.union(1, 2)
uf_demo.union(2, 3)
assert component_size(uf_demo, 1) == 3
```

---

## 2.6 本题小结

P3367 的价值在于把“图连通性”抽象成“集合合并与查询”。它本身不难，但会反复出现在其它算法中。

| 场景 | 并查集作用 |
|---|---|
| 动态加边查连通 | 判断两个点是否属于同一连通块 |
| Kruskal | 判断加入边是否成环 |
| 社交关系合并 | 维护群体大小、代表元 |
| 离线反向操作 | 把删边/删点转化为加边/加点 |

本题最应积累的是：**只要操作是“合并集合 + 查询同属关系”，先想并查集。**

# 题目三：P1113 杂务

[Luogu P1113](https://www.luogu.com.cn/problem/P1113)

## 3.1 题目说明

有 $N$ 项工作，每项工作 $i$ 需要耗时 $t_i$。某些工作必须在若干前置工作完成后才能开始。所有没有依赖冲突的工作可以并行进行，求完成全部工作的最短时间。

输入通常按工作编号给出：每一行包含工作编号、该工作耗时，以及若干前置工作编号，最后用 `0` 结束。

目标是输出所有工作全部完成的最早时间。

这不是普通的“把时间加起来”。如果两项工作互不依赖，它们可以同时进行；真正限制总时间的是依赖图上的最长链，也就是关键路径。


---

## 3.2 思路分析

把每项工作看成一个点。如果工作 $u$ 是工作 $v$ 的前置任务，就连一条有向边：

$$
u \to v
$$

由于“前置关系”不能自相矛盾，正常输入应形成 DAG（有向无环图）。设 $dp[i]$ 表示工作 $i$ 的最早完成时间。

如果 $i$ 没有前置任务：

$$
dp[i] = t_i
$$

如果 $i$ 有若干前驱 $pre$，则必须等所有前驱完成后才能开始：

$$
dp[i] = t_i + \max_{pre \to i} dp[pre]
$$

拓扑排序保证我们处理 $i$ 时，它的所有前驱都已经被处理过。答案为：

$$
\max_{1\le i\le N} dp[i]
$$

### 实现方式

1. 读入每项工作的前置任务 `pre`；
2. 对每个 `pre` 建边 `pre -> i`；
3. 入度为 $0$ 的工作可以立即开始，令 `dp[i] = duration[i]`；
4. BFS 式拓扑排序，沿边更新后继的最早完成时间。

复杂度：

$$
O(N+M)
$$

其中 $M$ 是依赖关系数量。


---

## 3.3 Python 参考代码

这一格是完整提交版。


```python
import sys
from collections import deque


def solve(data: str) -> str:
    lines = [line.strip() for line in data.splitlines() if line.strip()]
    n = int(lines[0])

    duration = [0] * (n + 1)
    graph = [[] for _ in range(n + 1)]
    indeg = [0] * (n + 1)

    for line in lines[1:]:
        nums = list(map(int, line.split()))
        job = nums[0]
        duration[job] = nums[1]
        for pre in nums[2:]:
            if pre == 0:
                break
            graph[pre].append(job)
            indeg[job] += 1

    dp = [0] * (n + 1)
    q = deque()
    for i in range(1, n + 1):
        if indeg[i] == 0:
            dp[i] = duration[i]
            q.append(i)

    processed = 0
    while q:
        u = q.popleft()
        processed += 1
        for v in graph[u]:
            dp[v] = max(dp[v], dp[u] + duration[v])
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)

    return str(max(dp[1:]))


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    print(solve(sys.stdin.read()))
```

---

## 3.4 测试验证

下面的依赖关系中，关键路径是：

$$
1 \to 2 \to 4 \to 6
$$

总耗时为：

$$
5+6+8+3=22
$$


```python
sample = """
6
1 5 0
2 6 1 0
3 4 1 0
4 8 2 3 0
5 2 2 0
6 3 4 5 0
"""

assert solve(sample) == "22"
print("P1113 测试通过：最短完成时间 =", solve(sample))
```

---

## 3.5 易错点与变体引出

这道题最关键的是方向：如果 `pre` 是 `job` 的前置任务，边应为 `pre -> job`，入度增加的是 `job`。方向反了，拓扑序和 DP 都会错。

还要注意：

1. **工作编号在输入中已经给出**：不要假设第几行就一定是几号工作，虽然很多数据看起来如此。
2. **多个前驱取最大，不是求和**：前驱可以并行完成，工作开始时间取最晚完成的那个前驱。
3. **答案是所有 `dp[i]` 的最大值**：不一定是编号最大的工作。
4. **拓扑排序只适用于 DAG**：若存在环，就表示依赖关系矛盾。

### 变体 3.5.1：课程安排能否完成 —— 拓扑排序判环

如果处理完的点数少于 $N$，说明图中存在环。此时不存在合法拓扑序。

**核心代码：**


```python
def has_cycle(n, graph, indeg):
    indeg = indeg[:]
    q = deque(i for i in range(1, n + 1) if indeg[i] == 0)
    count = 0

    while q:
        u = q.popleft()
        count += 1
        for v in graph[u]:
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)

    return count < n
```

### 变体 3.5.2：DAG 上最长路

P1113 的本质就是 DAG 上最长路。若每条边有权值 $w(u,v)$，可以写成：

$$
dp[v] = \max(dp[v],\ dp[u] + w(u,v))
$$

**核心代码：**


```python
def dag_longest_path(n, graph, indeg, start):
    """graph[u] = [(v, w), ...]；返回 start 到各点的最长距离。"""
    NEG_INF = -10**30
    indeg = indeg[:]
    dp = [NEG_INF] * (n + 1)
    dp[start] = 0
    q = deque(i for i in range(1, n + 1) if indeg[i] == 0)

    while q:
        u = q.popleft()
        for v, w in graph[u]:
            if dp[u] != NEG_INF:
                dp[v] = max(dp[v], dp[u] + w)
            indeg[v] -= 1
            if indeg[v] == 0:
                q.append(v)
    return dp
```

## 3.6 本题小结

P1113 是把“图”和“动态规划”连接起来的一道题。它告诉我们：当状态之间存在依赖关系时，如果依赖图是 DAG，就可以用拓扑序把递推顺序排出来。

| 关键词 | 对应思想 |
|---|---|
| 前置任务 | 有向边 |
| 可并行执行 | 不是简单求和 |
| 最短总工期 | 依赖图最长链 |
| 无前驱任务 | 拓扑排序起点 |
| 多个前驱 | 取最大完成时间 |

本题最应积累的是：**DAG 上的 DP，先找拓扑序，再写状态转移。**

# 题目四：P1144 最短路计数

[Luogu P1144](https://www.luogu.com.cn/problem/P1144)

## 4.1 题目说明

给定一张 $N$ 个点、$M$ 条边的无向无权图。要求对每个点 $i$，输出从 $1$ 到 $i$ 的最短路径条数，对 $100003$ 取模。

需要特别注意：图中可能存在重边。若两条边都能形成不同的最短路径，它们应分别计数。

输入为：

- 第一行：$N, M$
- 接下来 $M$ 行：$u, v$，表示一条无向边
- 输出 $N$ 行：第 $i$ 行输出从 $1$ 到 $i$ 的最短路径数量


---

## 4.2 思路分析

无权图的最短路由 BFS 决定。BFS 的层数就是距离：

$$
dist[v] = \text{从 }1\text{ 到 }v\text{ 的最少边数}
$$

同时维护 $cnt[v]$ 表示从 $1$ 到 $v$ 的最短路径条数。起点满足：

$$
dist[1]=0, \qquad cnt[1]=1
$$

从队首取出 $u$，遍历邻点 $v$ 时分两种情况：

### 第一次到达 $v$

如果 $dist[v] = -1$，说明还没访问过。此时经由 $u$ 到 $v$ 一定是最短路径：

$$
dist[v] = dist[u] + 1
$$

$$
cnt[v] = cnt[u]
$$

### 又发现一条等长最短路

如果：

$$
dist[v] = dist[u] + 1
$$

说明从 $u$ 接一条边到 $v$ 也是最短路径，于是累加：

$$
cnt[v] \leftarrow (cnt[v] + cnt[u]) \bmod 100003
$$

复杂度为 $O(N+M)$，空间复杂度也是 $O(N+M)$。


---

## 4.3 Python 参考代码

这一格是完整提交版。


```python
import sys
from collections import deque

MOD = 100003


def shortest_path_counts(n, graph, start=1):
    dist = [-1] * (n + 1)
    cnt = [0] * (n + 1)
    dist[start] = 0
    cnt[start] = 1
    q = deque([start])

    while q:
        u = q.popleft()
        for v in graph[u]:
            if dist[v] == -1:
                dist[v] = dist[u] + 1
                cnt[v] = cnt[u]
                q.append(v)
            elif dist[v] == dist[u] + 1:
                cnt[v] = (cnt[v] + cnt[u]) % MOD

    return cnt


def solve(data: str) -> str:
    it = iter(map(int, data.split()))
    n = next(it)
    m = next(it)
    graph = [[] for _ in range(n + 1)]

    for _ in range(m):
        u = next(it)
        v = next(it)
        graph[u].append(v)
        graph[v].append(u)

    cnt = shortest_path_counts(n, graph, 1)
    return "\n".join(str(cnt[i] % MOD) for i in range(1, n + 1))


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    print(solve(sys.stdin.read()))
```

## 4.5 易错点与变体引出

本题不是“统计所有路径”，而是“统计最短路径”。这意味着只能从第 $k$ 层走到第 $k+1$ 层时累加；如果边连向同层或更早层，都不能形成到该点的更短路径。

常见错误包括：

1. **忘记重边要计数**：邻接表中保留重边即可，自然会重复累加。
2. **把所有访问过的邻点都累加**：必须满足 `dist[v] == dist[u] + 1`。
3. **取模写错**：本题模数是 $100003$，不是常见的 $10^9+7$。
4. **误用 DFS**：无权最短路的层次需要 BFS 保证。

### 变体 4.5.1：带权图最短路计数

如果边有非负权，BFS 不再适用。可以在 Dijkstra 中同步维护 `cnt`：更短则覆盖，等长则累加。

**核心代码：**


```python
def count_shortest_paths_dijkstra(n, graph, start=1, mod=100003):
    INF = 10**30
    dist = [INF] * (n + 1)
    cnt = [0] * (n + 1)
    dist[start] = 0
    cnt[start] = 1
    heap = [(0, start)]

    while heap:
        d, u = heapq.heappop(heap)
        if d != dist[u]:
            continue
        for v, w in graph[u]:
            nd = d + w
            if nd < dist[v]:
                dist[v] = nd
                cnt[v] = cnt[u]
                heapq.heappush(heap, (nd, v))
            elif nd == dist[v]:
                cnt[v] = (cnt[v] + cnt[u]) % mod
    return dist, cnt
```

### 变体 4.5.2：固定最短路 DAG 后做 DP

BFS 求出 `dist` 后，只保留满足

$$
dist[v] = dist[u] + 1
$$

的边，就得到一张从低层指向高层的 DAG。之后可以按距离从小到大做路径计数 DP。

**核心代码：**


```python
def count_on_shortest_path_dag(n, graph, dist, start=1, mod=100003):
    order = sorted(range(1, n + 1), key=lambda x: dist[x] if dist[x] != -1 else 10**30)
    cnt = [0] * (n + 1)
    cnt[start] = 1

    for u in order:
        if dist[u] == -1:
            continue
        for v in graph[u]:
            if dist[v] == dist[u] + 1:
                cnt[v] = (cnt[v] + cnt[u]) % mod
    return cnt
```

## 4.6 本题小结

P1144 的核心收获是：**最短路不仅可以求距离，也可以在最短路结构上做计数。**

| 步骤 | 作用 |
|---|---|
| BFS 分层 | 确定每个点的最短距离 |
| 首次到达 | 继承前驱路径数 |
| 等长到达 | 累加前驱路径数 |
| 非等长边 | 忽略，不属于最短路贡献 |

本题是后续学习“最短路 DAG”“方案数 DP”的入口。看到“最短 + 计数”时，要自然想到：距离负责判定合法转移，计数负责累加方案。

# 题目五：P1330 封锁阳光大学

[Luogu P1330](https://www.luogu.com.cn/problem/P1330)

## 5.1 题目说明

给定一张无向图。我们需要在若干点上安排人员，使得每一条边的两个端点中，恰好有一个端点被安排。若无法做到，输出 `Impossible`；若可以做到，输出所需安排人数的最小值。

换一种图论语言：每条边都必须连接一个“选中点”和一个“未选中点”。因此相邻点必须属于不同阵营。这正是二分图染色问题。

对于每个连通块，如果可以二分染色，就有两种整体选择：选颜色 $0$ 的点，或选颜色 $1$ 的点。为了人数最少，对每个连通块取较小的一侧。


---

## 5.2 思路分析

设 `color[x]` 表示点 $x$ 的颜色，取值为 $0$ 或 $1$。对于每条边 $(u,v)$，必须满足：

$$
color[u] \ne color[v]
$$

如果在 BFS/DFS 染色过程中遇到一条边，其两个端点颜色相同，则图不是二分图，无解。

对一个连通块，假设染成两侧，大小分别为 $c_0,c_1$。该连通块的最优选择为：

$$
\min(c_0,c_1)
$$

整张图可能不连通，因此答案是所有连通块贡献之和：

$$
ans = \sum_{component} \min(c_0,c_1)
$$

如果某个连通块只有一个孤立点，没有边，那么它不需要覆盖任何道路，贡献应为 $0$。实现时可以通过“该连通块是否出现过边”来处理。

复杂度为 $O(N+M)$。


---

## 5.3 Python 参考代码

这一格是完整提交版。


```python
import sys
from collections import deque


def minimum_blockers(n, graph):
    color = [-1] * (n + 1)
    answer = 0

    for start in range(1, n + 1):
        if color[start] != -1:
            continue

        color[start] = 0
        count = [1, 0]
        has_edge = False
        q = deque([start])

        while q:
            u = q.popleft()
            if graph[u]:
                has_edge = True
            for v in graph[u]:
                if color[v] == -1:
                    color[v] = color[u] ^ 1
                    count[color[v]] += 1
                    q.append(v)
                elif color[v] == color[u]:
                    return None

        if has_edge:
            answer += min(count)

    return answer


def solve(data: str) -> str:
    it = iter(map(int, data.split()))
    n = next(it)
    m = next(it)
    graph = [[] for _ in range(n + 1)]

    for _ in range(m):
        u = next(it)
        v = next(it)
        graph[u].append(v)
        graph[v].append(u)

    ans = minimum_blockers(n, graph)
    return "Impossible" if ans is None else str(ans)


if __name__ == "__main__" and "ipykernel" not in sys.modules:
    print(solve(sys.stdin.read()))
```

## 5.4 测试验证

第一个样例是一个偶环，二分图成立，任选一侧两个点即可。第二个样例是三角形，奇环不可能二分染色，应输出 `Impossible`。


```python
possible = """
4 4
1 2
2 3
3 4
4 1
"""

impossible = """
3 3
1 2
2 3
3 1
"""

assert solve(possible) == "2"
assert solve(impossible) == "Impossible"
print("P1330 测试通过：")
print("偶环答案 =", solve(possible))
print("三角形答案 =", solve(impossible))
```

## 5.5 易错点与变体引出

这道题容易被误解成“一般图最小点覆盖”。一般图最小点覆盖很难，但本题多了“每条边两端必须一选一”的结构，等价于二分染色后在每个连通块中选择一侧。

需要注意：

1. **必须逐个连通块处理**：图不保证连通。
2. **不能全图只取一次颜色较小值**：每个连通块都可以独立翻转颜色，因此要逐块取 `min(c0,c1)`。
3. **奇环无解**：只要某个连通块染色冲突，整题就是 `Impossible`。
4. **孤立点贡献为 $0$**：没有边需要覆盖，不必安排人员。

### 变体 5.5.1：只判断一张图是否为二分图

如果题目只问能否二分染色，就不需要统计两侧大小，只要检查是否出现同色相邻。

**核心代码：**


```python
def is_bipartite(n, graph):
    color = [-1] * (n + 1)

    for start in range(1, n + 1):
        if color[start] != -1:
            continue
        color[start] = 0
        q = deque([start])
        while q:
            u = q.popleft()
            for v in graph[u]:
                if color[v] == -1:
                    color[v] = color[u] ^ 1
                    q.append(v)
                elif color[v] == color[u]:
                    return False
    return True
```

### 变体 5.5.2：树上最小点覆盖

如果图是一棵树，要选择尽量少的点覆盖所有边，可以做树形 DP。设：

- $dp[u][0]$：不选 $u$，覆盖 $u$ 子树内所有边的最小代价；
- $dp[u][1]$：选 $u$，覆盖 $u$ 子树内所有边的最小代价。

转移为：

$$
dp[u][0] = \sum_{v\in son(u)} dp[v][1]
$$

$$
dp[u][1] = 1 + \sum_{v\in son(u)} \min(dp[v][0], dp[v][1])
$$

**核心代码：**


```python
def tree_vertex_cover(n, tree, root=1):
    parent = [0] * (n + 1)
    order = [root]
    parent[root] = -1

    for u in order:
        for v in tree[u]:
            if v != parent[u]:
                parent[v] = u
                order.append(v)

    dp0 = [0] * (n + 1)
    dp1 = [1] * (n + 1)

    for u in reversed(order):
        for v in tree[u]:
            if parent[v] == u:
                dp0[u] += dp1[v]
                dp1[u] += min(dp0[v], dp1[v])

    return min(dp0[root], dp1[root])
```

## 5.6 本题小结

P1330 的核心是把题意翻译成“相邻点必须不同色”。一旦完成这一步，问题就变成标准二分图染色。

| 题意表达 | 图论翻译 |
|---|---|
| 每条边两端恰好选一个 | 相邻点颜色不同 |
| 无法安排 | 存在奇环，非二分图 |
| 人数最少 | 每个连通块选较小颜色类 |
| 图不连通 | 每个连通块独立处理 |

本题最应积累的是：**看到“相邻对象必须分成两类”，先想二分图；看到“奇环”，立刻想到二分图失败。**

5. **重视边界条件**：重边、孤立点、不连通图、堆中旧状态、多个前驱、奇环，这些往往决定能否 AC。
