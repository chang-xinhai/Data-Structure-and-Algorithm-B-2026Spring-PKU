# 题目一：约瑟夫问题

[Luogu P1996](https://www.luogu.com.cn/problem/P1996)



---

## 1.1 题目描述

> $n$ 个人围成一圈，从第 $1$ 个人开始报数。
> 数到 $m$ 的人出圈；
> 剩下的人从下一个位置继续从 $1$ 报数。
> 直到所有人都出圈为止。

**输出** 所有人出圈的顺序。

| 参数 | 范围 |
|------|------|
| $n$ | $1 \leq n \leq 100$ |
| $m$ | $1 \leq m \leq 100$ |

---

## 1.2 思路分析

### 为什么用队列来模拟？

围成一圈 $\Leftrightarrow$ **循环**。

**队列（Queue）** 的 FIFO 特性天然对应「按顺序报数」的行为：

1. 将所有人依次入队
2. 队首报数，数到 $m$ 的人出队（出圈）
3. 剩余的人从下一个位置继续从 $1$ 报数 $\Leftrightarrow$ 队首元素重新入队（排到队尾）
4. 循环直到队空

### 正确性直觉

> 每次报数的本质是把队首元素移到队尾重新排队。
> 当报到 $m$ 时，队首出队——正好对应出圈。
> 循环进行，最终出队顺序即为出圈顺序。

---

### 时间 · 空间复杂度

| 指标 | 复杂度 | 说明 |
|------|--------|------|
| **时间** | $O(n \times m)$ | 每次报数 $O(1)$，共约 $nm$ 次 |
| **空间** | $O(n)$ | 队列最多存 $n$ 个元素 |

此处 $n, m \leq 100$，完全够用。

---

## 1.3 参考代码

```python
from collections import deque

def josephus(n: int, m: int) -> list[int]:
    """队列模拟约瑟夫过程，返回出圈顺序"""
    q = deque(range(1, n + 1))   # 1 ~ n 入队
    order = []

    while q:
        # 报数：前 m-1 人移到队尾
        for _ in range(m - 1):
            q.append(q.popleft())
        # 第 m 个人出圈
        order.append(q.popleft())

    return order

n, m = map(int, input().split())
print(' '.join(map(str, josephus(n, m))))
```

# 1.4 变体讨论



### 变体 1.4.1：只求最后幸存者 — 数学 $O(n)$ 解法

当 $n$ 很大时（$n = 10^6$, $m = 20$），队列模拟会超时。

**递推思路**：

从 $0$ 开始编号（方便取模）。设 $f[i]$ 为「$i$ 个人围成一圈时最后幸存者的编号」，则有：

$$
f[1] = 0
\qquad
f[i] = \bigl(f[i-1] + m\bigr) \bmod i \quad (i \geq 2)
$$

**代码实现**：

```python
def josephus_math(n: int, m: int) -> int:
    """数学递推，O(n)，返回 1~n 编号的幸存者"""
    survivor = 0          # f[1] = 0（从 0 开始编号）
    for i in range(2, n + 1):
        survivor = (survivor + m) % i
    return survivor + 1   # 转回 1~n 编号

print(josephus_math(5, 2))   # → 3
```

### 变体 1.4.2：$m = 2$ 的特殊公式

当报数上限恒为 $2$ 时，存在更简洁的表达式：

$$
\text{ans} = 2 \times \Bigl(n - 2^{\lfloor \log_2 n \rfloor}\Bigr) + 1
$$

**代码实现**：

```python
def josephus_k2(n: int) -> int:
    """m=2 的特殊公式，O(1)"""
    k = 1 << (n.bit_length() - 1)   # 最大的 2^k <= n
    return 2 * (n - k) + 1

# 验证：对比公式与模拟
for n in range(1, 21):
    formula  = josephus_k2(n)
    simulate = josephus(n, 2)[-1]
    print(f"n={n:2d}  公式={formula:2d}  模拟={simulate:2d}  "
          f"{'✓' if formula == simulate else '✗'}")
```

### 变体 1.4.3：双向交替报数 — 循环链表 $O(n)$

**题目来源**：经典约瑟夫问题（OpenJudge xlxxsjjg/2026hw3/A）

> $N$ 个人站成一圈，顺时针编号 $1\sim N$。
> 从 $1$ 开始顺时针数到第 $K$ 个人出圈，
> 然后下一个人继续**从 $1$ 开始**顺时针数；
> 直至所有人出圈。

> $N \leq 2\times 10^6$，$K \leq 20$

**难点**：$N$ 太大，队列模拟超时——需要用**循环链表**实现，删除操作 $O(1)$。

用 `nxt[i]` 数组模拟指针：`nxt[i]` = 编号 $i$ 的下一个节点编号。

```python
def josephus_circle(n: int, k: int) -> list[int]:
    """数组模拟循环链表，O(n)"""
    nxt = list(range(n + 2))   # nxt[i] = i 的下一个编号
    nxt[n] = 1                 # 形成循环

    cur = 1
    order = []
    for _ in range(n):
        # 数 k-1 步
        for _ in range(k - 1):
            cur = nxt[cur]
        # cur 出圈
        order.append(cur)
        nxt[cur - 1] = nxt[cur]   # 从链表中删除 cur
        cur = nxt[cur]             # 移动到下一个

    return order

print(josephus_circle(7, 2))   # → [2, 4, 6, 1, 5, 3, 7]
```

# 题目二：后缀表达式求值

[Luogu P1449](https://www.luogu.com.cn/problem/P1449)



---

## 2.1 题目描述

> 给定一个**后缀表达式**（逆波兰表达式）。
> 数字以 `.` 结束，表达式以 `@` 结束。
> 运算符：`+`, `-`, `*`, `/`（除法向 0 取整）。
> 求表达式的整数值。

### 什么是后缀表达式？

| 中缀表达式 | 后缀表达式 |
|-----------|-----------|
| $3 + 4$ | `3 4 +` |
| $2 \times (3 + 4)$ | `2 3 4 + *` |
| $(1 + 2) \times 3 - 4$ | `1 2 + 3 * 4 -` |

**核心规则**：
- 遇到**数字** $\rightarrow$ 入栈
- 遇到**运算符** $\rightarrow$ 弹出两个数字，计算结果后入栈

---

## 2.2 思路分析

### 栈的本质作用

后缀表达式不需要括号，因为**运算符的优先级已经隐含在表达式的顺序中**。

栈的作用是**临时存储操作数，等待运算符来消耗它们**——这正是 LIFO 特性的体现。

### 手动推演（以 `6 5 2 3 + 8 * + 3 + *` 为例）

| 读取 | 操作 | 栈状态 |
|:----:|------|--------|
| `6` | 入栈 | `[6]` |
| `5` | 入栈 | `[6, 5]` |
| `2` | 入栈 | `[6, 5, 2]` |
| `3` | 入栈 | `[6, 5, 2, 3]` |
| `+` | 弹出 `3, 2`；算 $2+3=5$；入栈 | `[6, 5, 5]` |
| `8` | 入栈 | `[6, 5, 5, 8]` |
| `*` | 弹出 `8, 5`；算 $5\times8=40$；入栈 | `[6, 5, 40]` |
| `+` | 弹出 `40, 5`；算 $5+40=45$；入栈 | `[6, 45]` |
| `3` | 入栈 | `[6, 45, 3]` |
| `+` | 弹出 `3, 45`；算 $45+3=48$；入栈 | `[6, 48]` |
| `*` | 弹出 `48, 6`；算 $6\times48=288$；入栈 | `[288]` |

最终栈顶即为答案 $\Rightarrow 288$。

---

### 实现要点

1. 逐字符扫描，`while ch != '@'` 循环
2. 数字可能**多位**，用 `num = num * 10 + digit` 拼接
3. 遇到 `.` → 数字已在上面入栈，什么都不做
4. 遇到运算符时：**后弹出的是除数**，**先弹出的是被除数**
   `b = stack.pop()`，`a = stack.pop()`，算 $a \; op \; b$
5. Python 的 `//` 向负无穷取整，需用 `int(a / b)`（向 0 取整）

---

## 2.3 参考代码

```python
def evaluate_postfix(expr: str) -> int:
    """后缀表达式求值，运算符 + - * /，数字以 . 结束，@ 结束"""
    stack = []
    i = 0
    while i < len(expr) and expr[i] != '@':
        ch = expr[i]

        if ch.isdigit():
            num = 0
            while i < len(expr) and expr[i].isdigit():
                num = num * 10 + int(expr[i])
                i += 1
            stack.append(num)
            continue

        elif ch == '.':
            pass          # 数字已在上面入栈

        elif ch in '+-*/':
            b = stack.pop()   # 后弹出 → 除数（右侧操作数）
            a = stack.pop()   # 先弹出 → 被除数（左侧操作数）
            if   ch == '+': stack.append(a + b)
            elif ch == '-': stack.append(a - b)
            elif ch == '*': stack.append(a * b)
            elif ch == '/': stack.append(int(a / b))   # 向 0 取整

        i += 1

    return stack[-1]

expr = input().strip()
print(evaluate_postfix(expr))

# 示例：输入  6 5 2 3 + 8 * + 3 + *@
# 对应中缀:  6 * (5 + (2+3)*8 + 3) = 288
```

# 2.4 变体讨论



### 变体 2.4.1：带括号的中缀表达式求值 — 两个栈

**题目来源**：OpenJudge dsb12/26hwpractice/0404
**Luogu 类似**：[P1981](https://www.luogu.com.cn/problem/P1981)（NOIP 2013 普及组）

> 给定包含 `+ - * / ( )` 和整数的四则运算表达式，
> 计算结果，保留两位小数。

**解题思路：两个栈**

| 栈 | 存放 |
|----|------|
| **操作数栈** | 数字 |
| **运算符栈** | 运算符（含括号） |

**运算符优先级**（数越大优先级越高）：

| 运算符 | `(` | `+` `-` | `*` `/` | `)` |
|:------:|:---:|:-------:|:-------:|:---:|
| 优先级 | —（特殊） | $2$ | $3$ | —（特殊） |

**核心规则**：

1. 遇到**数字** $\rightarrow$ 入操作数栈
2. 遇到**运算符**：若栈顶优先级 $\geq$ 当前，**先弹出计算**
3. 遇到 `(` $\rightarrow$ 入运算符栈（作为特殊标记）
4. 遇到 `)` $\rightarrow$ 弹出运算符直到遇见匹配的 `(`
5. 表达式结束 $\rightarrow$ 弹出所有剩余运算符

```python
import re

def infix_eval(expr: str) -> float:
    def priority(op: str) -> int:
        if op in '+-': return 1
        if op in '*/': return 2
        return 0

    def calc(a: float, op: str, b: float) -> float:
        return {'+':a+b, '-':a-b, '*':a*b, '/':a/b}[op]

    nums, ops = [], []
    tokens = re.findall(r'\d+\.?\d*|[+\-*/()]', expr)

    for tok in tokens:
        if tok.lstrip('-').replace('.','').isdigit():
            nums.append(float(tok))
        elif tok == '(':
            ops.append(tok)
        elif tok == ')':
            while ops and ops[-1] != '(':
                nums.append(calc(nums.pop(), ops.pop(), nums.pop()))
            ops.pop()           # 弹出 '('
        else:                  # + - * /
            while (ops and ops[-1] != '('
                   and priority(ops[-1]) >= priority(tok)):
                nums.append(calc(nums.pop(), ops.pop(), nums.pop()))
            ops.append(tok)

    while ops:
        nums.append(calc(nums.pop(), ops.pop(), nums.pop()))
    return nums[-1]

print(f"3+4*2      = {infix_eval('3+4*2')}")
print(f"(3+4)*2    = {infix_eval('(3+4)*2')}")
print(f"3+4*(7+2)  = {infix_eval('3+4*(7+2)')}")
```

### 变体 2.4.2：前缀表达式（波兰表达式）求值

**规则**：运算符写在操作数前面。

| 中缀表达式 | 前缀表达式 |
|-----------|-----------|
| $3 + 4$ | `+ 3 4` |
| $2 \times (3 + 4)$ | `* 2 + 3 4` |

**求值方法**：从右往左扫描，规则与后缀完全对称。

```python
def evaluate_prefix(tokens: str) -> float:
    """前缀表达式求值：从右往左扫，数字入栈，运算符弹出两个计算后入栈"""
    stack = []
    tokens = tokens.split()[::-1]   # 反转，从右往左处理

    for tok in tokens:
        if tok.lstrip('-').isdigit():
            stack.append(float(tok))
        elif tok in '+-*/':
            a = stack.pop()
            b = stack.pop()
            stack.append({'+':a+b, '-':a-b, '*':a*b, '/':a/b}[tok])
    return stack[-1]

print(evaluate_prefix('+ 3 4'))        # → 7.0
print(evaluate_prefix('* + 3 4 2'))   # → 14.0
```

### 变体 2.4.3：表达式括号匹配

**Luogu**：[P1739](https://www.luogu.com.cn/problem/P1739)

> 给定以 `@` 结尾的表达式，判断其中的圆括号是否成对匹配。
> 匹配输出 `YES`，否则输出 `NO`。

这是栈的最基础应用——$O(n)$。

```python
def bracket_match(expr: str) -> str:
    stack = []
    for ch in expr:
        if ch == '(':
            stack.append(ch)
        elif ch == ')':
            if not stack or stack[-1] != '(':
                return 'NO'
            stack.pop()
    return 'YES' if not stack else 'NO'

print(bracket_match('(a+b)*(c+d)@'))    # YES
print(bracket_match('((a+b)*(c+d)@'))   # NO
```

# 本节小结



---

## 三种线性结构对比

| | 链表 | 栈 | 队列 |
|--|------|----|------|
| **操作口** | 任意位置 O(1) | 一端 push/pop | 两端 enq/deq |
| **特性** | 指针串联 | LIFO（后进先出）| FIFO（先进先出） |
| **实现** | 指针或数组 | 数组或链表 | 数组或链表 |
| **典型应用** | 约瑟夫、邻接表 | 表达式求值、DFS | BFS、任务调度 |

---

## 什么时候用什么结构？

| 需求 | 推荐结构 | 原因 |
|------|----------|------|
| 频繁在**中间**插入/删除 | 链表 | O(1) 指针修改 |
| 需要「**最近**先处理」（函数调用、DFS）| 栈 | LIFO 天然匹配 |
| 需要「**先来**先服务」（BFS、任务队列）| 队列 | FIFO 天然匹配 |
| 约瑟夫问题 | 循环链表或队列 | 循环结构的模拟 |

---

## 约瑟夫问题的核心洞见

> **模拟** vs **数学**：
> - 模拟（队列 / 链表）：直观、工程实用，但大数据下可能 $O(n\times m)$ 超时
> - 数学递推：$O(n)$ 高效，但需要推导能力
> - 两种思路都重要：面试常考递推，工程中模拟更常见

> **链表 vs 数组**：
> - 链表插入 / 删除 $O(1)$，但无法随机访问
> - 数组随机访问 $O(1)$，但插入 / 删除需要移动元素
> - 选择取决于「读」多还是「写」多

---

## 表达式求值的核心洞见

> 栈天然适合「**嵌套**」结构：
> - 后缀表达式：运算符等待右侧两个操作数就绪
> - 中缀表达式：括号改变了默认优先级
> - 函数调用：参数等待函数体计算完毕

> 本质都是 **LIFO**——后进的操作数先被消耗。
