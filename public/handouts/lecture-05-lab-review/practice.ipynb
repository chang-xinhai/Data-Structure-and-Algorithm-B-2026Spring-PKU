## A. 句子反转

题目链接：[A:句子反转](http://dsb12.openjudge.cn/26preexam/A/)

**题目概述**

输入包含多行英文句子，每行不含标点。对每一行，保持每个单词内部字符不变，只反转单词顺序。

- 输入：若干行，每行一个英文句子；
- 输出：每行输出单词顺序反转后的句子。


**思路**

**考点分析 / 知识点回忆**

- 考点：EOF 输入、逐行处理、字符串拆分与拼接。
- 对应知识点：一行就是一条记录；每次读一行，再把这一行里的单词反过来。
- 题型信号：题面出现“输入包含多行”“每行一个句子”。
- 常见错误：把所有单词一次性拆开，导致原来的行边界丢失。

这道题的关键不是算法，而是不要把多行输入混在一起。

最直观的写法是：不断 `input()`，读到 EOF 就停。每一行单独处理：

1. `line.split()` 得到这一行的单词；
2. `words[::-1]` 反转单词顺序；
3. `' '.join(...)` 拼回一行。

复杂度为 $O(L)$，其中 $L$ 是输入总字符数。


```python
# 写法 1：课堂推荐，逐行读到 EOF
while True:
    try:
        line = input()
    except EOFError:
        break

    words = line.split()
    print(' '.join(words[::-1]))

# 写法 2：OJ 中也常见，一次读完所有行
# import sys
# for line in sys.stdin.readlines():
#     words = line.split()
#     print(' '.join(words[::-1]))
```

**分析 / 变体讨论**

若题目改为“反转每个单词内部字符，但保持单词顺序”，则反转对象从单词列表变成每个单词本身。


```python
# 变体：保持单词顺序，反转每个单词内部字符
line = "I love PKU"
print(' '.join(word[::-1] for word in line.split()))
# 输出：I evol UKP
```

## B. 斐波那契数

题目链接：[B:斐波那契数](http://dsb12.openjudge.cn/26preexam/B/)

**题目概述**

给定多次询问，每次给出 $n$，输出 Fibonacci 数 $F(n)$。

$$
F(0)=0,\quad F(1)=1,\quad F(n)=F(n-1)+F(n-2).
$$

数据范围：$n\le 30$。


**思路**

**考点分析 / 知识点回忆**

- 考点：递推、预处理、多次询问。
- 对应知识点：Fibonacci 是最基础的重叠子问题；数组递推比朴素递归稳定。
- 题型信号：多组询问、范围很小、递推式已给出。
- 常见错误：每个询问重新递归计算，导致重复子问题爆炸。

不要对每个询问都递归计算。因为朴素递归会重复计算大量子问题，复杂度接近指数级。

由于最大 $n$ 很小，可以一次预处理：

$$
\text{fib}[i]=\text{fib}[i-1]+\text{fib}[i-2].
$$

之后每个询问 $O(1)$ 输出。


```python
t = int(input())

fib = [0] * 31
fib[1] = 1
for i in range(2, 31):
    fib[i] = fib[i - 1] + fib[i - 2]

for _ in range(t):
    n = int(input())
    print(fib[n])
```

**分析 / 变体讨论**

如果 $n$ 很大但只需要对某个模数取模，只需在递推时取模即可。


```python
# 变体：F(n) mod MOD
MOD = 10**9 + 7
n = 10
fib = [0] * (n + 1)
if n >= 1:
    fib[1] = 1
for i in range(2, n + 1):
    fib[i] = (fib[i - 1] + fib[i - 2]) % MOD
print(fib[n])
```

## C. stone game

题目链接：[C:stone game](http://dsb12.openjudge.cn/26preexam/C/)

**题目概述**

有 $n$ 个石头，Alice 与 Bob 轮流操作。每次选择一个正整数 $x$，满足 $x<n$ 且 $x\mid n$，然后移除 $x$ 个石头。无法操作者输。Alice 先手，双方都最优，判断胜者。

数据范围：$n<100000$。


**思路**

**考点分析 / 知识点回忆**

- 考点：博弈 DP、必胜/必败态、找规律。
- 对应知识点：若存在一步走到必败态，则当前为必胜态；若所有合法后继都是必胜态，则当前为必败态。
- 题型信号：两人轮流、双方最优、无法操作者输。
- 常见错误：只模拟 Alice 的贪心选择，而没有从状态胜负角度分析。

这是经典 Divisor Game。

结论：Alice 当且仅当 $n$ 为偶数时获胜。

证明思路：

- 若 $n$ 为偶数，Alice 可以取 $x=1$，把局面变成奇数；
- 若当前 $n$ 为奇数，则其所有因子 $x$ 都是奇数，移除后 $n-x$ 为偶数；
- 因此偶数局面可以交给对手奇数局面，奇数局面一定会交给对手偶数局面；
- $n=1$ 是必败态。

所以：

$$
win(n)= [n\equiv 0\pmod 2].
$$


```python
n = int(input())
print("Alice" if n % 2 == 0 else "Bob")
```

**分析 / 变体讨论**

如果不先看出奇偶规律，也可以用 DP 验证小范围状态：

$$
dp[i]=\exists x\mid i,\; 1\le x<i,\; dp[i-x]=False.
$$

这有助于从实验中发现规律。


```python
# 变体：用 DP 验证前若干项胜负
N = 12
dp = [False] * (N + 1)
for i in range(2, N + 1):
    for x in range(1, i):
        if i % x == 0 and not dp[i - x]:
            dp[i] = True
            break
print(["A" if dp[i] else "B" for i in range(1, N + 1)])
```

## D. 结点数量最多的层所在深度

题目链接：[D:结点数量最多的层所在深度](http://dsb12.openjudge.cn/26preexam/D/)

**题目概述**

给定若干棵二叉树。每棵树节点编号从 $1$ 开始，输入每个节点的左孩子和右孩子编号，空孩子用 $0$ 表示。根节点为 $1$。输出结点数量最多的一层的深度；若多层并列，输出深度最小者。深度从 $0$ 开始。


**思路**

**考点分析 / 知识点回忆**

- 考点：二叉树存储、层序遍历、BFS 层计数。
- 对应知识点：层相关问题优先 BFS；队列中保存节点及其深度。
- 题型信号：题面问“第几层”“层所在深度”“层次遍历”。
- 常见错误：深度从 0 还是 1 开始看错；并列最大时没有选择最小深度。

层相关问题优先使用 BFS。

队列中保存二元组 $(u,d)$，表示节点 $u$ 位于深度 $d$。维护数组 $cnt[d]$ 统计每层节点数。

BFS 不变量：队列中每个节点都带着正确深度；当节点弹出时，立即计入对应层。

最后选择：

$$
\arg\max_d cnt[d],
$$

若并列，保留最小 $d$。


```python
from collections import deque

T = int(input())

for _ in range(T):
    n = int(input())
    children = [(0, 0)] * (n + 1)
    for i in range(1, n + 1):
        left, right = map(int, input().split())
        children[i] = (left, right)

    q = deque([(1, 0)])
    cnt = []
    while q:
        u, d = q.popleft()
        if d == len(cnt):
            cnt.append(0)
        cnt[d] += 1
        for v in children[u]:
            if v != 0:
                q.append((v, d + 1))

    best = max(range(len(cnt)), key=lambda d: cnt[d])
    print(best)
```

**分析 / 变体讨论**

若题目要求输出最大宽度本身，而不是层深度，只需输出 $\max_d cnt[d]$。


```python
# 变体：给定每层节点数 cnt，求最大宽度
cnt = [1, 2, 4, 3]
print(max(cnt))
```

## E. 图像渲染

题目链接：[E:图像渲染](http://dsb12.openjudge.cn/26preexam/E/)

**题目概述**

给定一张 $m\times n$ 图像，从起点 $(sr,sc)$ 开始，把与起点原颜色相同且四方向连通的区域染成新颜色 `color`。有多组测试数据。


**思路**

**考点分析 / 知识点回忆**

- 考点：Flood Fill、矩阵 BFS/DFS、四方向扩展。
- 对应知识点：连通块搜索；BFS 入队即标记，或 DFS 递归标记。
- 题型信号：从某个像素开始，对上下左右相邻且同色的区域执行操作。
- 常见错误：没有先保存原颜色；新颜色等于旧颜色时死循环或重复处理。

这是 Flood Fill。

关键步骤：

1. 记录原颜色 $old=image[sr][sc]$；
2. 若 $old=color$，直接输出原图；
3. 从起点 BFS；
4. 每次扩展四邻居，只有颜色仍为 $old$ 的格子才染色并入队。

不变量：队列中的所有格子都已经被染成新颜色，且属于起点所在的原色连通块。


```python
from collections import deque

T = int(input())

for _ in range(T):
    m, n = map(int, input().split())
    sr, sc, color = map(int, input().split())
    image = [list(map(int, input().split())) for _ in range(m)]

    old = image[sr][sc]
    if old != color:
        q = deque([(sr, sc)])
        image[sr][sc] = color
        while q:
            x, y = q.popleft()
            for dx, dy in ((1, 0), (-1, 0), (0, 1), (0, -1)):
                nx, ny = x + dx, y + dy
                if 0 <= nx < m and 0 <= ny < n and image[nx][ny] == old:
                    image[nx][ny] = color
                    q.append((nx, ny))

    for row in image:
        print(*row)
```

**分析 / 变体讨论**

若题目只要求连通块大小，不要求修改图像，可以把“染色”改为访问标记并计数。


```python
# 变体：统计从起点出发的同色连通块大小
from collections import deque
image = [[1, 1, 0], [1, 0, 0], [1, 1, 1]]
m, n = 3, 3
old = image[0][0]
seen = {(0, 0)}
q = deque([(0, 0)])
while q:
    x, y = q.popleft()
    for dx, dy in ((1,0),(-1,0),(0,1),(0,-1)):
        nx, ny = x + dx, y + dy
        if 0 <= nx < m and 0 <= ny < n and (nx, ny) not in seen and image[nx][ny] == old:
            seen.add((nx, ny))
            q.append((nx, ny))
print(len(seen))
```

## F. 成绩排序

题目链接：[F:成绩排序](http://dsb12.openjudge.cn/26preexam/F/)

**题目概述**

有 $n$ 名同学，成绩互不相同。给出 $m$ 个询问回答：对 $x<y$，回答 $z=1$ 表示 $x$ 比 $y$ 高，$z=0$ 表示 $x$ 不比 $y$ 高，即 $y$ 比 $x$ 高。问有多少名同学的精确排名可以确定。


**思路**

**考点分析 / 知识点回忆**

- 考点：有向图、传递闭包、排名可确定性。
- 对应知识点：Floyd 的布尔版本可以求任意两点是否可达。
- 题型信号：关系具有传递性，问能否推断、能确定多少排名。
- 常见错误：把题目当成普通排序；只看直接入度/出度，忽略间接关系。

这不是排序题，而是有向关系推断。

建图：若 $a$ 比 $b$ 高，则加边 $a\to b$。通过传递性求闭包。

Floyd 后，$reach[i][j]$ 表示可以推出 $i$ 比 $j$ 高。

某个同学 $i$ 的排名可确定，当且仅当它和其他所有同学的相对关系都能确定：

$$
\sum_{j\ne i} [reach[i][j]\lor reach[j][i]] = n-1.
$$

$n\le 50$，使用 $O(n^3)$ Floyd 最稳。


```python
n, m = map(int, input().split())
reach = [[False] * n for _ in range(n)]

for _ in range(m):
    x, y, z = map(int, input().split())
    if z == 1:
        reach[x][y] = True
    else:
        reach[y][x] = True

for k in range(n):
    for i in range(n):
        if reach[i][k]:
            for j in range(n):
                reach[i][j] = reach[i][j] or reach[k][j]

ans = 0
for i in range(n):
    known = 0
    for j in range(n):
        if i != j and (reach[i][j] or reach[j][i]):
            known += 1
    if known == n - 1:
        ans += 1
print(ans)
```

**分析 / 变体讨论**

若题目问“能否确定完整排名”，则需要每个人的排名都可确定，即答案是否等于 $n$。


```python
# 变体：已知 ans 和 n，判断完整排名是否唯一
n = 4
ans = 4
print("unique" if ans == n else "not unique")
```

## A. 5月14日

题目链接：[A:5月14日](http://xlxxsjjg.openjudge.cn/2026midexam/A/)

**题目概述**

给定整数 $m,n$，求闭区间 $[m,n]$ 中所有能被 $19$ 整除的数之和。


**思路**

**考点分析 / 知识点回忆**

- 考点：基础循环、条件判断、累加器。
- 对应知识点：闭区间枚举 `range(m, n + 1)`；整除判断 `x % 19 == 0`。
- 题型信号：范围很小，要求统计/求和满足条件的数。
- 常见错误：漏掉端点；把 `< n` 写成 `<= n` 的相反错误。

直接枚举即可。范围 $0<m<n<1000$，复杂度 $O(n-m+1)$ 足够。

也可以用等差数列求和，但在上机第一题中直接循环更稳。


```python
m, n = map(int, input().split())
ans = 0
for x in range(m, n + 1):
    if x % 19 == 0:
        ans += x
print(ans)
```

**分析 / 变体讨论**

若范围很大，可以求第一个与最后一个 $19$ 的倍数，再用等差数列求和。


```python
# 变体：等差数列求 [m,n] 中 19 的倍数和
m, n = 50, 95
first = ((m + 18) // 19) * 19
last = (n // 19) * 19
if first > last:
    print(0)
else:
    cnt = (last - first) // 19 + 1
    print((first + last) * cnt // 2)
```

## B. 云服务

题目链接：[B:云服务](http://xlxxsjjg.openjudge.cn/2026midexam/B/)

**题目概述**

有 $n$ 类任务。第 $i$ 类任务共有 $a_i$ 个，专用节点最多免费处理 $r_i$ 个，超出的任务需要交给公用云，每个费用 $c$。求最小总费用。


**思路**

**考点分析 / 知识点回忆**

- 考点：线性扫描、逐项贡献、业务题转数学表达式。
- 对应知识点：总量由独立单项贡献相加；每类任务互不影响。
- 题型信号：每一类都有容量和需求，问最小总费用。
- 常见错误：模拟任务分配过程，而不是直接计算超出容量的部分。

每类任务互相独立。第 $i$ 类需要交给公用云的数量为：

$$
\max(0,a_i-r_i).
$$

所以总费用为：

$$
 c\sum_{i=1}^n \max(0,a_i-r_i).
$$

复杂度 $O(n)$。


```python
n, c = map(int, input().split())
r = list(map(int, input().split()))
a = list(map(int, input().split()))
extra = sum(max(0, a[i] - r[i]) for i in range(n))
print(extra * c)
```

**分析 / 变体讨论**

若公用云对不同类型任务费用不同，只需把单价 $c$ 换成 $c_i$。


```python
# 变体：每类任务单价不同
r = [3, 2, 5]
a = [4, 1, 8]
price = [7, 10, 2]
print(sum(max(0, need - cap) * p for need, cap, p in zip(a, r, price)))
```

## C. 索菲的血瓶

题目链接：[C:索菲的血瓶](http://xlxxsjjg.openjudge.cn/2026midexam/C/)

**题目概述**

给定小灯墙高度序列 $a_1,\dots,a_n$ 和索菲墙高度序列 $b_1,\dots,b_w$。索菲墙可以整体升高或降低任意常数。问小灯墙中有多少个长度为 $w$ 的连续区间与索菲墙相同，允许整体平移。

$n,w\le 2\times 10^5$。


**思路**

**考点分析 / 知识点回忆**

- 考点：差分序列、字符串匹配、KMP。
- 对应知识点：整体平移不改变相邻差分；KMP 可在线性时间统计模式串出现次数。
- 题型信号：允许整体升高/降低，问连续区间匹配个数。
- 常见错误：直接比较原高度；忘记处理 `w=1` 的特殊情况。

两个长度为 $w$ 的序列允许整体平移匹配，当且仅当它们的相邻差分完全相同：

$$
a_i-b_i=C \iff a_{i+1}-a_i=b_{i+1}-b_i.
$$

因此将原问题转化为：在 $a$ 的差分序列中匹配 $b$ 的差分序列。

若 $w=1$，任意单个瓶塔都可以通过平移匹配，答案为 $n$。

差分匹配可以用 KMP，复杂度 $O(n+w)$。


```python
def prefix_function(p):
    pi = [0] * len(p)
    for i in range(1, len(p)):
        j = pi[i - 1]
        while j > 0 and p[i] != p[j]:
            j = pi[j - 1]
        if p[i] == p[j]:
            j += 1
        pi[i] = j
    return pi

n, w = map(int, input().split())
a = list(map(int, input().split()))
b = list(map(int, input().split()))

if w == 1:
    print(n)
else:
    da = [a[i + 1] - a[i] for i in range(n - 1)]
    db = [b[i + 1] - b[i] for i in range(w - 1)]
    pi = prefix_function(db)
    j = 0
    ans = 0
    for x in da:
        while j > 0 and x != db[j]:
            j = pi[j - 1]
        if x == db[j]:
            j += 1
        if j == len(db):
            ans += 1
            j = pi[j - 1]
    print(ans)
```

**分析 / 变体讨论**

若不允许整体升降，则直接在原序列上做模式匹配。若允许整体乘以同一比例，则需要比较相邻比值，但要注意除法精度，应改用交叉相乘。


```python
# 变体：朴素检查某个窗口是否可整体平移匹配
window = [4, 5, 5, 4, 3]
pattern = [2, 3, 3, 2, 1]
shift = window[0] - pattern[0]
print(all(x - y == shift for x, y in zip(window, pattern)))
```

## D. 队列pk

题目链接：[D:队列pk](http://xlxxsjjg.openjudge.cn/2026midexam/D/)

**题目概述**

给定 $1\sim n$ 的一个排列，$n$ 为偶数。每次从当前队列中取出两个相邻整数，按原顺序追加到新队列末尾。所有数取完后，希望最终新队列字典序最大。


**思路**

**考点分析 / 知识点回忆**

- 考点：贪心、双向链表、堆维护候选。
- 对应知识点：字典序最大先保证第一个位置最大；删除相邻元素后需要快速找到新的相邻关系。
- 题型信号：反复从序列中删除相邻块，并要求最终字典序最优。
- 常见错误：用列表中间删除导致 $O(n^2)$；只按原始相邻关系判断，删除后没有更新。

每一步输出的下一个数一定是当前队列中某条相邻边的左端点。因此所有“当前不是最后一个”的元素都可以作为下一输出位置的候选。

为使字典序最大，应选择当前仍有右邻居的最大元素 $u$，并将 $(u,next[u])$ 这一对取出。由于排列元素互不相同，第一关键字已经决定最优。

用双向链表维护当前队列，用最大堆维护仍可能作为左端点的元素。删除一对后，只需要把新的前驱重新加入堆。

复杂度 $O(n\log n)$。


```python
import heapq

n = int(input())
a = [0] + list(map(int, input().split()))

prev = [i - 1 for i in range(n + 2)]
nxt = [i + 1 for i in range(n + 2)]
nxt[n] = 0
alive = [True] * (n + 1)

heap = []
for i in range(1, n + 1):
    if nxt[i] != 0:
        heapq.heappush(heap, (-a[i], i))

ans = []
for _ in range(n // 2):
    while True:
        _, i = heapq.heappop(heap)
        j = nxt[i]
        if alive[i] and j != 0 and alive[j]:
            break

    ans.extend([a[i], a[j]])
    L, R = prev[i], nxt[j]
    alive[i] = alive[j] = False
    if L != 0:
        nxt[L] = R
    if R != 0:
        prev[R] = L
    if L != 0 and R != 0:
        heapq.heappush(heap, (-a[L], L))

print(*ans)
```

**分析 / 变体讨论**

若元素不互异，则堆的关键字需要比较整对 $(a_i,a_{next_i})$，因为第一关键字可能相同。


```python
# 变体：若元素可重复，候选相邻对可按 pair 比较
pairs = [(4, 2), (4, 3), (3, 9)]
print(max(pairs))  # Python 元组按字典序比较
```

## E. 算力集群部署

题目链接：[E:算力集群部署](http://xlxxsjjg.openjudge.cn/2026midexam/E/)

**题目概述**

给定目标字符串，每次可以选择一个连续区间，把这个区间全部覆盖成同一种字符，后操作会覆盖前操作。求得到目标字符串所需的最少操作次数。$n\le 50$。


**思路**

**考点分析 / 知识点回忆**

- 考点：区间 DP、覆盖操作、后操作覆盖前操作。
- 对应知识点：区间 DP 通常按长度从小到大枚举；连续重复字符可压缩。
- 题型信号：每次操作一个连续区间，要求最少操作次数。
- 常见错误：贪心按字符段数计算，忽略后续覆盖可以合并同字符操作。

这是典型区间 DP，类似 Strange Printer。

令 $dp[l][r]$ 表示打印出子串 $s[l:r+1]$ 的最少次数。

基础转移：先单独打印 $s[l]$，再打印后面：

$$
dp[l][r]\le 1+dp[l+1][r].
$$

如果存在 $k$ 使得 $s[k]=s[l]$，那么 $s[l]$ 可以和 $s[k]$ 在同一次操作中完成：

$$
dp[l][r]\le dp[l+1][k-1]+dp[k][r].
$$

连续重复字符可先压缩，不影响答案。


```python
s = input().strip()
# 压缩连续重复字符
compressed = []
for ch in s:
    if not compressed or compressed[-1] != ch:
        compressed.append(ch)
s = ''.join(compressed)

n = len(s)
dp = [[0] * n for _ in range(n)]
for i in range(n):
    dp[i][i] = 1

for length in range(2, n + 1):
    for l in range(0, n - length + 1):
        r = l + length - 1
        dp[l][r] = dp[l + 1][r] + 1
        for k in range(l + 1, r + 1):
            if s[k] == s[l]:
                mid = dp[l + 1][k - 1] if l + 1 <= k - 1 else 0
                dp[l][r] = min(dp[l][r], mid + dp[k][r])

print(dp[0][n - 1] if n else 0)
```

**分析 / 变体讨论**

若每次只能覆盖长度固定的区间，则状态需要加入长度限制；若覆盖不能重叠，则会变成区间划分问题。


```python
# 变体：压缩连续重复字符
s = "AAABBBAAC"
compressed = []
for ch in s:
    if not compressed or compressed[-1] != ch:
        compressed.append(ch)
print(''.join(compressed))
```

## F. 二叉树作业

题目链接：[F:二叉树作业](http://xlxxsjjg.openjudge.cn/2026midexam/F/)

**题目概述**

二叉树用字符串编码：`0` 表示叶子，`1S` 表示一个孩子，`2S1S2` 表示两个孩子。用红、绿、蓝三色染色，父子颜色不同；若有两个孩子，两个孩子颜色也不同。求合法染色中最多蓝色节点数与最少绿色节点数。

字符串长度可达 $5\times 10^5$。


**思路**

**考点分析 / 知识点回忆**

- 考点：树编码解析、树形 DP、颜色状态。
- 对应知识点：先序编码可用栈或递归解析；树形 DP 需要枚举根节点颜色与孩子颜色。
- 题型信号：树由字符串递归定义，目标是最大/最小某种颜色数量。
- 常见错误：递归深度过大；没有区分“最多蓝色”和“最少绿色”是两个不同 DP 目标。

先根据先序编码建出树。由于规模大，使用迭代建树并按编号逆序做 DP。

对每个节点 $u$ 和颜色 $c$，维护：

- $mx[u][c]$：当 $u$ 染成 $c$ 时，子树最多蓝色数；
- $mn[u][c]$：当 $u$ 染成 $c$ 时，子树最少绿色数。

叶子：

$$
mx[u][c]=[c=Blue],\quad mn[u][c]=[c=Green].
$$

一个孩子：孩子颜色不能等于父亲颜色。

两个孩子：父亲和两个孩子三者颜色必须两两不同。


```python
S = input().strip()
n = len(S)
children = [[] for _ in range(n)]
need = []
stack = []

for idx, ch in enumerate(S):
    arity = int(ch)
    need.append(arity)
    if stack:
        parent = stack[-1]
        children[parent].append(idx)
        need[parent] -= 1
        while stack and need[stack[-1]] == 0:
            stack.pop()
    if arity > 0:
        stack.append(idx)

mx = [[0] * 3 for _ in range(n)]  # 0 red, 1 green, 2 blue
mn = [[0] * 3 for _ in range(n)]

for u in range(n - 1, -1, -1):
    if not children[u]:
        for c in range(3):
            mx[u][c] = 1 if c == 2 else 0
            mn[u][c] = 1 if c == 1 else 0
    elif len(children[u]) == 1:
        v = children[u][0]
        for c in range(3):
            mx[u][c] = (1 if c == 2 else 0) + max(mx[v][d] for d in range(3) if d != c)
            mn[u][c] = (1 if c == 1 else 0) + min(mn[v][d] for d in range(3) if d != c)
    else:
        a, b = children[u]
        for c in range(3):
            other = [d for d in range(3) if d != c]
            x, y = other
            mx[u][c] = (1 if c == 2 else 0) + max(mx[a][x] + mx[b][y], mx[a][y] + mx[b][x])
            mn[u][c] = (1 if c == 1 else 0) + min(mn[a][x] + mn[b][y], mn[a][y] + mn[b][x])

print(max(mx[0]), min(mn[0]))
```

**分析 / 变体讨论**

若颜色数从 $3$ 扩展到 $K$，一个节点有两个孩子时就需要枚举两个不同的子颜色，复杂度会多一个 $K^2$ 因子。


```python
# 变体：枚举父色 c 下两个孩子的不同颜色
K = 4
c = 0
pairs = [(a, b) for a in range(K) for b in range(K) if a != c and b != c and a != b]
print(pairs)
```

## G. 数据采集

题目链接：[G:数据采集](http://xlxxsjjg.openjudge.cn/2026midexam/G/)

**题目概述**

无人机从 $(1,1)$ 出发，每一步横坐标加一。状态包含高度 $y$、垂直速度 $v$、离开基准层的连续时间。每步可选择加速度 $-1,0,1$，到达新格子后可能受到风力扰动或获得奖励。要求到达 $(m,1)$ 的最大累计奖励。

$n\le100,m\le1000,k\le10$。


**思路**

**考点分析 / 知识点回忆**

- 考点：状态 DP、滚动数组、网格状态建模。
- 对应知识点：影响未来决策的信息必须进入状态；横坐标每步加一，所以可以按列推进。
- 题型信号：题面包含位置、速度、时间限制、奖励最大化。
- 常见错误：状态只记位置，漏掉速度或离线时间；边界截断和基准层重置处理不一致。

这是状态 DP。影响未来的量不只有位置，还包括速度与离线时间。

令状态为：

$$
(x,y,v,t),
$$

其中 $t$ 表示连续离开基准层的时间。因为每一步 $x$ 必定加一，可以按列滚动 DP，只保存当前列状态。

转移：选择 $a\in\{-1,0,1\}$，先令 $v'=v+a$，再到达 $y'=y+v'$。若超过上边界，截断到 $n$；若低于 $1$，该转移非法。到达 $y'=1$ 时速度重置为 $0$，离线时间重置为 $0$。


```python
n, m, k = map(int, input().split())
raw = [input().split() for _ in range(n)]  # 输入从高层到低层
cell = [[None] * (m + 1) for _ in range(n + 1)]
for row in range(n):
    y = n - row
    for x in range(1, m + 1):
        token = raw[row][x - 1]
        cell[y][x] = (token[0], int(token[1:]))

NEG = -10**18
V_LIMIT = n + 20 * k + 5
cur = {(1, 0, 0): 0}  # (高度, 速度, 离线时间) -> 已得收益

for x in range(1, m):
    nxt = {}
    for (y, v, t), val in cur.items():
        for acc in (-1, 0, 1):
            nv = v + acc
            ny = y + nv
            if ny > n:
                ny = n
            if ny < 1:
                continue

            if ny == 1:
                state = (1, 0, 0)
                gain = 0
            else:
                nt = t + 1
                if nt > k:
                    continue
                typ, num = cell[ny][x + 1]
                if typ == 'f':
                    nv2 = nv + num
                    if abs(nv2) > V_LIMIT:
                        continue
                    state = (ny, nv2, nt)
                    gain = 0
                else:
                    state = (ny, nv, nt)
                    gain = num

            if val + gain > nxt.get(state, NEG):
                nxt[state] = val + gain
    cur = nxt

print(cur.get((1, 0, 0), NEG))
```

**分析 / 变体讨论**

若没有速度和离线时间约束，问题会退化为普通网格 DP；若横坐标不固定增加，则会变成一般图上的最短路/最长路问题，需要重新考虑环。


```python
# 变体：普通从左到右网格最大收益 DP
n, m = 2, 3
reward = [[0, 1, 2], [0, 5, 1]]
dp = [[-10**9] * m for _ in range(n)]
dp[0][0] = 0
for x in range(m - 1):
    for y in range(n):
        for ny in range(n):
            dp[ny][x + 1] = max(dp[ny][x + 1], dp[y][x] + reward[ny][x + 1])
print(max(dp[y][m - 1] for y in range(n)))
```

## C. 二叉树修剪

题目链接：[C:二叉树修剪](http://xlxxsjjg.openjudge.cn/2026midsim/C/)

**题目概述**

给定一棵以 $1$ 为根的二叉树，每条边有苹果数。要求保留恰好 $Q$ 条树枝，使保留下来的苹果数最大。

$N\le100$。


**思路**

**考点分析 / 知识点回忆**

- 考点：树形 DP、树上背包、子树合并。
- 对应知识点：`dp[u][k]` 表示子树内使用 k 条边的最大收益；合并孩子时做背包。
- 题型信号：树上保留固定数量边/点，最大化权值收益。
- 常见错误：把每个子树独立取最大，忽略总边数需要在孩子之间分配。

树形背包。

令 $dp[u][k]$ 表示在 $u$ 的子树中保留 $k$ 条边时，最多能保留多少苹果。

合并一个孩子 $v$，边权为 $w$ 时，若给该孩子子树分配 $t$ 条边，则还要额外保留边 $(u,v)$，贡献为：

$$
dp_v[t]+w.
$$

转移：

$$
new[i+t+1]=\max(new[i+t+1], dp_u[i]+dp_v[t]+w).
$$


```python
import sys
sys.setrecursionlimit(10000)

N, Q = map(int, input().split())
tree = [[] for _ in range(N + 1)]
for _ in range(N - 1):
    parent, child, apples = map(int, input().split())
    tree[parent].append((child, apples))

NEG = -10**9
dp = [[NEG] * (Q + 1) for _ in range(N + 1)]
size = [0] * (N + 1)

def dfs(u):
    dp[u][0] = 0
    size[u] = 0
    for v, w in tree[u]:
        dfs(v)
        new = [NEG] * (Q + 1)
        for i in range(0, min(size[u], Q) + 1):
            if dp[u][i] <= NEG:
                continue
            for t in range(0, min(size[v], Q - i - 1) + 1):
                new[i + t + 1] = max(new[i + t + 1], dp[u][i] + dp[v][t] + w)
            new[i] = max(new[i], dp[u][i])
        size[u] += size[v] + 1
        dp[u] = new

dfs(1)
print(dp[1][Q])
```

**分析 / 变体讨论**

若题目要求“最多保留 $Q$ 条边”，最后取 $\max_{0\le k\le Q} dp[1][k]$。


```python
# 变体：恰好 Q 改为最多 Q
dp_root = [0, 5, 9, 8]
Q = 3
print(max(dp_root[:Q + 1]))
```

## D. 只要功夫深，猴子当文豪

题目链接：[D:只要功夫深，猴子当文豪](http://xlxxsjjg.openjudge.cn/2026midsim/D/)

**题目概述**

猴子随机敲 $S$ 次键盘，键盘上有 $K$ 个按键，每个按键是一个大写字母。目标单词长度为 $L$。目标单词可重叠出现。若准备“最大可能出现次数”根香蕉，问吃完实际出现次数后剩余香蕉数的期望。


**思路**

**考点分析 / 知识点回忆**

- 考点：KMP 前后缀、概率期望、重叠匹配。
- 对应知识点：期望的线性性；最大重叠次数由最长公共前后缀决定。
- 题型信号：目标串可能重叠出现，问期望或最多出现次数。
- 常见错误：把出现事件当成互斥；最大次数没有考虑重叠。

答案为：

$$
\text{max_occurrence}-\mathbb E[\text{occurrence}].
$$

若目标中有字符不在键盘上，则最大出现次数和期望都为 $0$。

期望出现次数：每个起点出现目标串的概率相同，为目标每个字符概率的乘积。因此：

$$
\mathbb E=(S-L+1)\prod_{ch\in target}P(ch).
$$

最大出现次数由目标串的最长公共前后缀决定。若最小平移为 $shift$，则：

$$
1+\left\lfloor\frac{S-L}{shift}\right\rfloor.
$$


```python
from collections import Counter

def prefix_function(s):
    pi = [0] * len(s)
    for i in range(1, len(s)):
        j = pi[i - 1]
        while j and s[i] != s[j]:
            j = pi[j - 1]
        if s[i] == s[j]:
            j += 1
        pi[i] = j
    return pi

T = int(input())
for case in range(1, T + 1):
    K, L, S = map(int, input().split())
    keyboard = input().strip()
    target = input().strip()
    cnt = Counter(keyboard)

    if S < L or any(ch not in cnt for ch in target):
        ans = 0.0
    else:
        prob = 1.0
        for ch in target:
            prob *= cnt[ch] / K
        expected = (S - L + 1) * prob

        pi = prefix_function(target)
        shift = L - pi[-1]
        max_occ = 1 + (S - L) // shift
        ans = max_occ - expected

    print(f"Case #{case}: {ans:.3f}")
```

**分析 / 变体讨论**

若只问目标串在给定文本中出现多少次，则直接使用 KMP 计数即可。


```python
# 变体：KMP 统计文本中模式串出现次数
text = "ABABA"
pat = "ABA"
combined = pat + '#' + text
pi = prefix_function(combined)
print(sum(1 for x in pi if x == len(pat)))
```

## E. D++

题目链接：[E:D++](http://xlxxsjjg.openjudge.cn/2026midsim/E/)

**题目概述**

给出若干段只包含 `F i x y` 与 `E` 的伪代码，以及声称的复杂度 $O(1)$ 或 $O(n^w)$。判断实际复杂度是否一致，若语法错误输出 `ERR`。


**思路**

**考点分析 / 知识点回忆**

- 考点：栈模拟、作用域、复杂度指数统计、错误检测。
- 对应知识点：嵌套结构用栈；当前复杂度指数等于有效嵌套中贡献 n 的层数。
- 题型信号：伪代码、循环嵌套、变量作用域、要求判断复杂度。
- 常见错误：不会执行的循环内部仍要检查语法错误；变量出作用域时忘记删除。

这是栈模拟。

维护：

- 当前活跃变量集合；
- 循环栈；
- 当前有效嵌套中贡献 $n$ 的层数；
- 最大贡献层数。

一个循环是否贡献 $n$：当父层有效，且下界为常数、上界为 $n$ 时，贡献 $1$。

循环是否执行也要判断。例如 $F i n 1$ 不执行，但其中变量重复等语法错误仍需检查。


```python
T = int(input())

def parse_complexity(s):
    if s == 'O(1)':
        return 0
    if s == 'O(n)':
        return 1
    return int(s.split('^')[1][:-1])

def greater(x, y):
    if x == 'n' and y != 'n':
        return True
    if x != 'n' and y != 'n' and int(x) > int(y):
        return True
    return False

for _ in range(T):
    L, comp = input().split()
    L = int(L)
    target = parse_complexity(comp)

    stack = []  # (变量名, 是否贡献复杂度, 当前循环是否能执行)
    active_vars = set()
    cur_exp = 0
    max_exp = 0
    err = False

    for _ in range(L):
        parts = input().split()
        if parts[0] == 'F':
            _, var, x, y = parts
            if var in active_vars:
                err = True
            active_vars.add(var)

            parent_active = stack[-1][2] if stack else True
            loop_active = parent_active and not greater(x, y)
            contributes = loop_active and x != 'n' and y == 'n'
            if contributes:
                cur_exp += 1
                max_exp = max(max_exp, cur_exp)
            stack.append((var, contributes, loop_active))
        else:
            if not stack:
                err = True
            else:
                var, contributes, _ = stack.pop()
                active_vars.remove(var)
                if contributes:
                    cur_exp -= 1

    if stack:
        err = True

    if err:
        print('ERR')
    else:
        print('Yes' if max_exp == target else 'No')
```

**分析 / 变体讨论**

若增加 `while` 或 `if` 语句，仍然先确定每种语句对“作用域、是否执行、复杂度指数”的影响，再加入栈或状态机。


```python
# 变体：解析复杂度字符串
def parse_complexity(s):
    if s == 'O(1)':
        return 0
    if s == 'O(n)':
        return 1
    return int(s.split('^')[1][:-1])

for s in ['O(1)', 'O(n^3)']:
    print(parse_complexity(s))
```

## F. 无尽迷宫

题目链接：[F:无尽迷宫](http://xlxxsjjg.openjudge.cn/2026midsim/F/)

**题目概述**

给定一个 $N\times M$ 的迷宫单元，它在无限平面上周期重复。问从起点 `S` 出发，能否走到任意远处。


**思路**

**考点分析 / 知识点回忆**

- 考点：BFS、周期网格、无限状态压缩。
- 对应知识点：真实坐标可以无限，但地形由模坐标决定；同一模坐标若以不同真实坐标到达，则可逃向无限远。
- 题型信号：无限迷宫、周期重复、问能否走到任意远。
- 常见错误：只在基本块内 BFS，无法判断跨周期逃逸。

在无限坐标上 BFS，但访问状态按模周期归类。

若同一个模位置 $(x\bmod N,y\bmod M)$ 被不同的真实坐标到达，说明可以通过周期位移走向无限远，答案为 `Yes`。

否则 BFS 最终会停在有限状态，答案为 `No`。


```python
from collections import deque

def can_escape(n, m, grid):
    sx = sy = 0
    for r in range(n):
        for c, ch in enumerate(grid[r]):
            if ch == 'S':
                sx, sy = r, c

    q = deque([(sx, sy)])
    seen = {(sx % n, sy % m): (sx, sy)}

    while q:
        x, y = q.popleft()
        for dx, dy in ((1, 0), (-1, 0), (0, 1), (0, -1)):
            nx, ny = x + dx, y + dy
            rx, ry = nx % n, ny % m
            if grid[rx][ry] == '#':
                continue
            key = (rx, ry)
            if key in seen:
                if seen[key] != (nx, ny):
                    return True
            else:
                seen[key] = (nx, ny)
                q.append((nx, ny))
    return False

while True:
    try:
        line = input().strip()
    except EOFError:
        break

    if line == '':
        continue
    n, m = map(int, line.split())
    grid = [input().strip() for _ in range(n)]

    print('Yes' if can_escape(n, m, grid) else 'No')
```

**分析 / 变体讨论**

普通有限迷宫只需要记录 $(x,y)$ 是否访问过；周期无限迷宫需要额外比较“同一模位置的真实坐标是否不同”。


```python
# 变体：同一模位置出现不同真实坐标，说明存在周期位移
n, m = 3, 3
p1 = (1, 1)
p2 = (4, 1)
print((p1[0] % n, p1[1] % m) == (p2[0] % n, p2[1] % m), p1 != p2)
```

## G. 量子态制备序列

题目链接：[G:量子态制备序列](http://xlxxsjjg.openjudge.cn/2026midsim/G/)

**题目概述**

有 $N$ 个 0/1 位。每次操作必须恰好翻转 $M$ 个不同位置。给定初态、终态和操作次数 $K$，求恰好 $K$ 次操作后从初态到终态的操作序列数，模 $10^9+7$。


**思路**

**考点分析 / 知识点回忆**

- 考点：组合计数、按汉明距离 DP、状态压缩。
- 对应知识点：只关心当前有多少位与目标不同，不关心具体哪些位；转移用组合数计数。
- 题型信号：每次翻转固定数量元素，问经过 K 步的序列数。
- 常见错误：直接对 $2^N$ 个状态 DP；没有把状态压缩为错误位数量。

只关心当前有多少位与目标不同。设当前不同位数量为 $d$。

一次操作中，若从这 $d$ 个错误位里选 $x$ 个翻转，它们会变正确；同时从 $N-d$ 个正确位中选 $M-x$ 个翻转，它们会变错误。新错误数：

$$
d'=d-x+(M-x)=d+M-2x.
$$

转移方案数：

$$
\binom{d}{x}\binom{N-d}{M-x}.
$$

初始错误数为 $0$，目标错误数为初态与终态的汉明距离 $H$。做 $K$ 步 DP。


```python
MOD = 10**9 + 7
N, K, M = map(int, input().split())
start = list(map(int, input().split()))
target = list(map(int, input().split()))
H = sum(a != b for a, b in zip(start, target))

# 组合数
C = [[0] * (N + 1) for _ in range(N + 1)]
for i in range(N + 1):
    C[i][0] = C[i][i] = 1
    for j in range(1, i):
        C[i][j] = (C[i - 1][j - 1] + C[i - 1][j]) % MOD

dp = [0] * (N + 1)
dp[0] = 1
for _ in range(K):
    ndp = [0] * (N + 1)
    for d in range(N + 1):
        if dp[d] == 0:
            continue
        lo = max(0, M - (N - d))
        hi = min(d, M)
        for x in range(lo, hi + 1):
            nd = d + M - 2 * x
            ways = C[d][x] * C[N - d][M - x]
            ndp[nd] = (ndp[nd] + dp[d] * ways) % MOD
    dp = ndp

print(dp[H])
```

**分析 / 变体讨论**

如果每次操作可以翻转“至多 $M$ 个位置”，则需要把本题转移中的固定 $M$ 改成枚举 $0\le t\le M$。


```python
# 变体：计算初态与终态的汉明距离
start = [0, 1, 0, 1]
target = [1, 1, 0, 0]
print(sum(a != b for a, b in zip(start, target)))
```
