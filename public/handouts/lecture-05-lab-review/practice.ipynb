## A. 句子反转

题目链接：[A:句子反转](http://dsb12.openjudge.cn/26preexam/A/)

**题目概述**

输入包含多行英文句子，每行不含标点。对每一行，保持每个单词内部字符不变，只反转单词顺序。

- 输入：若干行，每行一个英文句子；
- 输出：每行输出单词顺序反转后的句子。


**思路**

**考点分析 / 知识点回忆**

- 考点：EOF 输入、逐行处理、字符串拆分与拼接。
- 对应知识点：一行就是一条记录；每次读一行，再把这一行里的单词反过来。
- 题型信号：题面出现“输入包含多行”“每行一个句子”。
- 常见错误：把所有单词一次性拆开，导致原来的行边界丢失。

这道题的关键不是算法，而是不要把多行输入混在一起。

最直观的写法是：不断 `input()`，读到 EOF 就停。每一行单独处理：

1. `line.split()` 得到这一行的单词；
2. `words[::-1]` 反转单词顺序；
3. `' '.join(...)` 拼回一行。

复杂度为 $O(L)$，其中 $L$ 是输入总字符数。

**多种解法路线**

- 解法 1：`input()` 逐行读到 EOF。最适合课堂讲，结构清楚。时间复杂度 $O(L)$；空间复杂度 $O(\ell_{\max})$，其中 $L$ 是总字符数，$\ell_{\max}$ 是最长行长度。
- 解法 2：一次读完所有行再逐行处理。OJ 里也常见，但不要用 `split()` 把行边界弄丢。时间复杂度 $O(L)$；空间复杂度 $O(L)$。

这题不是算法难，而是输入格式容易写歪：一行是一条记录，就必须一行一行反转。


```python

# 写法 1：课堂推荐，逐行读到 EOF
while True:                         # 不知道有多少行，所以一直读
    try:                            # 尝试读取一整行
        line = input()              # line 保留这一行的边界
    except EOFError:                # 读到文件结束时，input 会抛出 EOFError
        break                       # 没有更多输入，退出循环

    words = line.split()            # 把当前行拆成单词列表
    print(' '.join(words[::-1]))    # 反转单词顺序，再用空格拼回字符串

# 写法 2：OJ 中也常见，一次读完所有行
# import sys
# for line in sys.stdin.readlines(): # readlines() 保留“按行”的结构
#     words = line.split()           # 只拆当前这一行，不打乱其他行
#     print(' '.join(words[::-1]))   # 输出这一行反转后的结果
```

**分析 / 变体讨论**

若题目改为“反转每个单词内部字符，但保持单词顺序”，则反转对象从单词列表变成每个单词本身。


```python

# 变体：保持单词顺序，反转每个单词内部字符
line = "I love PKU"                                      # 示例句子
words = line.split()                                      # 拆成 ['I', 'love', 'PKU']
reversed_words = [word[::-1] for word in words]           # 每个单词内部反转
print(' '.join(reversed_words))                           # 单词顺序不变，拼回一句话
# 输出：I evol UKP
```

## B. 斐波那契数

题目链接：[B:斐波那契数](http://dsb12.openjudge.cn/26preexam/B/)

**题目概述**

给定多次询问，每次给出 $n$，输出 Fibonacci 数 $F(n)$。

$$
F(0)=0,\quad F(1)=1,\quad F(n)=F(n-1)+F(n-2).
$$

数据范围：$n\le 30$。


**思路**

**考点分析 / 知识点回忆**

- 考点：递推、预处理、多次询问。
- 对应知识点：Fibonacci 是最基础的重叠子问题；数组递推比朴素递归稳定。
- 题型信号：多组询问、范围很小、递推式已给出。
- 常见错误：每个询问重新递归计算，导致大量重复子问题。

不要对每个询问都直接展开递归。朴素递归的调用次数满足 Fibonacci 型递推，可以精确估到 $\Theta(\varphi^n)$。

由于最大 $n$ 很小，可以一次预处理：

$$
\text{fib}[i]=\text{fib}[i-1]+\text{fib}[i-2].
$$

之后每个询问 $O(1)$ 输出。

**多种解法路线**

- 解法 1：朴素递归。照着定义写 `fib(n-1)+fib(n-2)`，最好理解，但会重复计算同一批子问题，适合用来引出记忆化。设最大询问为 $N$、询问数为 $T$，时间复杂度 $O(T\varphi^N)$；空间复杂度 $O(N)$，其中 $\varphi=\frac{1+\sqrt5}{2}$。
- 解法 2：记忆化搜索。递归不变，加一个表保存已经算过的值。时间复杂度 $O(N+T)$；空间复杂度 $O(N)$。
- 解法 3：递推预处理。考试里最稳：先把 `fib[0..30]` 算好，多次询问直接输出。时间复杂度 $O(N+T)$；空间复杂度 $O(N)$。

可以先给学生看朴素递归的坑，再把“重复子问题”自然引到 DP：


```python

# 朴素递归：演示单个 n；n 稍大时会重复计算很多子问题
n = int(input())                     # 读入要求的 Fibonacci 下标

def fib(k):                          # 返回 F(k)
    if k <= 1:                       # 递归边界：F(0)=0, F(1)=1
        return k
    return fib(k - 1) + fib(k - 2)   # 按定义递归计算 F(k-1)+F(k-2)

print(fib(n))                        # 输出 F(n)
```

```python

# 记忆化搜索：仍然按递归定义写，但每个状态只算一次
n = int(input())                     # 读入要求的 Fibonacci 下标
memo = [-1] * (n + 1)                # memo[k] 存 F(k)，-1 表示还没算过

def fib(k):                          # 返回 F(k)
    if k <= 1:                       # 递归边界
        return k
    if memo[k] != -1:                # 已经算过，直接复用答案
        return memo[k]
    memo[k] = fib(k - 1) + fib(k - 2) # 第一次遇到 k，递归计算并保存
    return memo[k]                   # 返回保存好的结果

print(fib(n))                        # 输出 F(n)
```

```python

t = int(input())                     # 询问次数

fib = [0] * 31                       # 题目范围 n <= 30，预留 0..30
fib[1] = 1                           # F(1)=1，F(0) 默认是 0
for i in range(2, 31):               # 从小到大递推
    fib[i] = fib[i - 1] + fib[i - 2] # F(i)=F(i-1)+F(i-2)

for _ in range(t):                   # 逐个回答询问
    n = int(input())                 # 当前询问的 n
    print(fib[n])                    # O(1) 输出预处理结果
```

**分析 / 变体讨论**

如果 $n$ 很大但只需要对某个模数取模，只需在递推时取模即可。


```python

# 变体：F(n) mod MOD
MOD = 10**9 + 7                      # 常见取模数
n = 10                               # 示例下标
fib = [0] * (n + 1)                  # fib[i] 存 F(i) mod MOD
if n >= 1:                           # 避免 n=0 时访问 fib[1] 越界
    fib[1] = 1
for i in range(2, n + 1):            # 递推每一项
    fib[i] = (fib[i - 1] + fib[i - 2]) % MOD # 每一步都取模，防止数变得过大
print(fib[n])                        # 输出答案
```

## C. stone game

题目链接：[C:stone game](http://dsb12.openjudge.cn/26preexam/C/)

**题目概述**

有 $n$ 个石头，Alice 与 Bob 轮流操作。每次选择一个正整数 $x$，满足 $x<n$ 且 $x\mid n$，然后移除 $x$ 个石头。无法操作者输。Alice 先手，双方都最优，判断胜者。

数据范围：$n<100000$。


**思路**

**考点分析 / 知识点回忆**

- 考点：博弈搜索、博弈 DP、必胜/必败态、找规律。
- 对应知识点：若存在一步走到必败态，则当前为必胜态；若所有合法后继都是必胜态，则当前为必败态。
- 题型信号：两人轮流、双方最优、无法操作者输。
- 常见错误：一上来背结论，或者只模拟 Alice 贪心取最大因子。

这题课堂上不要直接跳到“偶数必胜”。更自然的推进是：先会搜索，再会 DP，最后从 DP 表里看出数学规律。

**多种解法路线**

**解法 1：搜索所有走法（朴素博弈树）**

把 `win(n)` 定义成：当前面对 $n$ 个石头的人是否必胜。枚举所有合法操作 $x$，只要存在一种操作能让对手进入必败态，当前就是必胜。若不记忆化，递归会重复展开同一状态；按代码枚举因子时，每层还有 $O(k)$ 检查，时间复杂度可用递推 $T(k)=O(k)+\sum_{x\mid k}T(k-x)$ 描述；空间复杂度为递归深度 $O(n)$。


```python

# 只能用于很小的 n，用来理解状态转移
n = int(input())                     # 当前石子数量

def win(k):                          # 当前面对 k 个石头的人是否必胜
    for x in range(1, k):            # 枚举可以取走的数量 x
        if k % x == 0 and not win(k - x): # x 是 k 的因子，且能把对手送到必败态
            return True              # 存在一种赢法，当前状态必胜
    return False                     # 所有后继都让对手必胜，当前状态必败

print("Alice" if win(n) else "Bob") # Alice 先手，判断初始状态胜负
```



这个写法很直观，但它按博弈树重复展开同一状态；同一个 $k-x$ 会在不同路径中反复出现，因此适合先理解状态转移，不适合作为最终提交写法。

**解法 2：DP 记住每个状态（可过很多小数据，用来找规律）**

把递归改成从小到大填表。若枚举 $x=1,2,\dots,i-1$ 检查因子，总检查次数为 $\sum_{i=2}^{n}(i-1)=O(n^2)$，所以时间复杂度 $O(n^2)$；空间复杂度 $O(n)$。

$$
dp[i]=\exists x,\, x\mid i,\,1\le x<i,\, dp[i-x]=False.
$$


```python

n = int(input())                     # 初始石子数量
dp = [False] * (n + 1)               # dp[i] 表示面对 i 个石头时先手是否必胜

for i in range(2, n + 1):            # 从小状态推到大状态；dp[1] 默认 False
    for x in range(1, i):            # 枚举本轮取走的 x
        if i % x == 0 and not dp[i - x]: # 如果能走到一个对手必败态
            dp[i] = True             # 当前状态就是必胜态
            break                    # 找到一种赢法即可，不必继续枚举

print("Alice" if dp[n] else "Bob")  # 输出先手 Alice 是否获胜
```



这比搜索好，因为每个状态只算一次。它的价值是帮助我们打表观察：

| n | 1 | 2 | 3 | 4 | 5 | 6 | 7 | 8 | ... |
|---|---|---|---|---|---|---|---|---|---|
| win | B | A | B | A | B | A | B | A | ... |

**解法 3：数学规律（最终推荐）**

结论：Alice 当且仅当 $n$ 为偶数时获胜。

证明很短：

- $n=1$ 时没有合法操作，先手必败；
- 若 $n$ 为偶数，Alice 可以取 $x=1$，把局面变成奇数；
- 若 $n$ 为奇数，则它的所有因子 $x$ 都是奇数，取走后 $n-x$ 一定是偶数；
- 所以偶数局面能交给对手奇数局面，奇数局面只能交给对手偶数局面。

因此最终只需要判断奇偶。时间复杂度 $O(1)$；空间复杂度 $O(1)$。

$$
win(n)= [n\equiv 0\pmod 2].
$$


```python

n = int(input())                     # 初始石子数量
print("Alice" if n % 2 == 0 else "Bob") # 偶数先手胜，奇数先手败
```

**分析 / 变体讨论**

如果不先看出奇偶规律，也可以用 DP 验证小范围状态：

$$
dp[i]=\exists x\mid i,\; 1\le x<i,\; dp[i-x]=False.
$$

这有助于从实验中发现规律。


```python

# 变体：用 DP 验证前若干项胜负
N = 12                               # 只打表观察 1..12
dp = [False] * (N + 1)               # dp[i] 表示 i 个石头时先手是否必胜
for i in range(2, N + 1):            # 从小到大填表
    for x in range(1, i):            # 枚举合法取法
        if i % x == 0 and not dp[i - x]: # 能走到必败态，则当前必胜
            dp[i] = True
            break
print(["A" if dp[i] else "B" for i in range(1, N + 1)]) # A/B 展示规律
```

## D. 结点数量最多的层所在深度

题目链接：[D:结点数量最多的层所在深度](http://dsb12.openjudge.cn/26preexam/D/)

**题目概述**

给定若干棵二叉树。每棵树节点编号从 $1$ 开始，输入每个节点的左孩子和右孩子编号，空孩子用 $0$ 表示。根节点为 $1$。输出结点数量最多的一层的深度；若多层并列，输出深度最小者。深度从 $0$ 开始。


**思路**

**考点分析 / 知识点回忆**

- 考点：二叉树存储、层序遍历、BFS 层计数。
- 对应知识点：层相关问题优先 BFS；队列中保存节点及其深度。
- 题型信号：题面问“第几层”“层所在深度”“层次遍历”。
- 常见错误：深度从 0 还是 1 开始看错；并列最大时没有选择最小深度。

层相关问题优先使用 BFS。

队列中保存二元组 $(u,d)$，表示节点 $u$ 位于深度 $d$。维护数组 $cnt[d]$ 统计每层节点数。

BFS 不变量：队列中每个节点都带着正确深度；当节点弹出时，立即计入对应层。

最后选择：

$$
\arg\max_d cnt[d],
$$

若并列，保留最小 $d$。

**多种解法路线**

- 解法 1：BFS 层序遍历。最贴题，因为问题直接问“哪一层最多”。设节点数为 $n$、最大层宽为 $w$，时间复杂度 $O(n)$；队列空间 $O(w)$，若计入存树则空间 $O(n)$。
- 解法 2：DFS 带深度。递归/栈遍历每个节点，访问到深度 `d` 就让 `cnt[d] += 1`。设树高为 $h$，时间复杂度 $O(n)$；递归栈和层计数空间 $O(h)$，若计入存树则空间 $O(n)$。

DFS 写法也很短，适合作为 BFS 的对照：


```python

# DFS 带深度：完整可运行版本
import sys
sys.setrecursionlimit(1_000_000)     # 树可能较深，调高递归深度限制

T = int(input())                     # 测试数据组数

for _ in range(T):                   # 逐组处理
    n = int(input())                 # 当前树的节点数
    children = [(0, 0)] * (n + 1)    # children[i] = (左孩子, 右孩子)
    for i in range(1, n + 1):        # 读入每个节点的孩子
        left, right = map(int, input().split())
        children[i] = (left, right)

    cnt = []                         # cnt[d] 表示深度 d 的节点数

    def dfs(u, d):                   # 当前访问节点 u，深度为 d
        if u == 0:                   # 0 表示空孩子
            return
        if d == len(cnt):            # 第一次到达这个深度
            cnt.append(0)            # 新增一层计数
        cnt[d] += 1                  # 当前节点计入深度 d
        dfs(children[u][0], d + 1)   # 递归访问左子树
        dfs(children[u][1], d + 1)   # 递归访问右子树

    dfs(1, 0)                        # 根节点编号为 1，深度从 0 开始
    best = max(range(len(cnt)), key=lambda d: cnt[d]) # max 默认保留最小深度的并列答案
    print(best)                      # 输出节点数最多的层的深度
```



本题推荐 BFS，因为“层”这个信息在队列里天然存在；如果树很深，递归 DFS 还要注意递归深度。


```python

from collections import deque        # deque 用来做 BFS 队列

T = int(input())                     # 测试数据组数

for _ in range(T):                   # 逐组处理
    n = int(input())                 # 当前树的节点数
    children = [(0, 0)] * (n + 1)    # children[i] = (左孩子, 右孩子)
    for i in range(1, n + 1):        # 读入每个节点的孩子
        left, right = map(int, input().split())
        children[i] = (left, right)

    q = deque([(1, 0)])              # 队列元素是 (节点编号, 深度)
    cnt = []                         # cnt[d] 统计深度 d 的节点数
    while q:                         # 队列非空就继续层序遍历
        u, d = q.popleft()           # 取出一个节点及其深度
        if d == len(cnt):            # 第一次遇到这一层
            cnt.append(0)
        cnt[d] += 1                  # 当前节点计入这一层
        for v in children[u]:        # 枚举左右孩子
            if v != 0:               # 0 是空孩子，不入队
                q.append((v, d + 1)) # 孩子的深度是 d+1

    best = max(range(len(cnt)), key=lambda d: cnt[d]) # 并列时返回最小深度
    print(best)                      # 输出答案
```

**分析 / 变体讨论**

若题目要求输出最大宽度本身，而不是层深度，只需输出 $\max_d cnt[d]$。


```python

# 变体：给定每层节点数 cnt，求最大宽度
cnt = [1, 2, 4, 3]                   # cnt[d] 表示第 d 层节点数
print(max(cnt))                      # 最大层宽就是 cnt 的最大值
```

## E. 图像渲染

题目链接：[E:图像渲染](http://dsb12.openjudge.cn/26preexam/E/)

**题目概述**

给定一张 $m\times n$ 图像，从起点 $(sr,sc)$ 开始，把与起点原颜色相同且四方向连通的区域染成新颜色 `color`。有多组测试数据。


**思路**

**考点分析 / 知识点回忆**

- 考点：Flood Fill、矩阵 BFS/DFS、四方向扩展。
- 对应知识点：连通块搜索；BFS 入队即标记，或 DFS 递归标记。
- 题型信号：从某个像素开始，对上下左右相邻且同色的区域执行操作。
- 常见错误：没有先保存原颜色；新颜色等于旧颜色时死循环或重复处理。

这是 Flood Fill。

关键步骤：

1. 记录原颜色 $old=image[sr][sc]$；
2. 若 $old=color$，直接输出原图；
3. 从起点 BFS；
4. 每次扩展四邻居，只有颜色仍为 $old$ 的格子才染色并入队。

不变量：队列中的所有格子都已经被染成新颜色，且属于起点所在的原色连通块。

**多种解法路线**

- 解法 1：BFS，用队列一圈圈染色。优点是额外空间由队列显式控制，不依赖 Python 递归栈。设同色连通块大小为 $C$，搜索时间 $O(C)$；输出整张图后总时间 $O(mn)$；空间复杂度 $O(C)$。
- 解法 2：DFS 递归，代码更短，但图像很大时 Python 可能递归过深。搜索时间 $O(C)$；输出后总时间 $O(mn)$；递归栈空间 $O(C)$。

DFS 版本可以这样理解：当前位置如果不是旧颜色就停，否则先染色，再递归四个方向。


```python

# DFS Flood Fill：完整可运行版本
import sys
sys.setrecursionlimit(1_000_000)     # 连通块很大时递归可能很深

T = int(input())                     # 测试数据组数

for _ in range(T):                   # 逐组处理
    m, n = map(int, input().split()) # 图像行数 m、列数 n
    sr, sc, color = map(int, input().split()) # 起点坐标和新颜色
    image = [list(map(int, input().split())) for _ in range(m)] # 读入图像矩阵

    old = image[sr][sc]              # 起点的原颜色，只能染这个连通块

    def dfs(x, y):                   # 尝试染色坐标 (x, y)
        if x < 0 or x >= m or y < 0 or y >= n: # 越界就停止
            return
        if image[x][y] != old:       # 不是原颜色，不能属于目标连通块
            return
        image[x][y] = color          # 先染色，避免之后重复访问
        dfs(x + 1, y)                # 向下扩展
        dfs(x - 1, y)                # 向上扩展
        dfs(x, y + 1)                # 向右扩展
        dfs(x, y - 1)                # 向左扩展

    if old != color:                 # 新旧颜色相同则无需搜索，避免重复递归
        dfs(sr, sc)                  # 从起点开始染色

    for row in image:                # 输出染色后的图像
        print(*row)
```



如果 `old == color`，两种写法都要直接结束，否则会反复访问同一片区域。


```python

from collections import deque        # BFS 队列

T = int(input())                     # 测试数据组数

for _ in range(T):                   # 逐组处理
    m, n = map(int, input().split()) # 图像大小
    sr, sc, color = map(int, input().split()) # 起点和目标颜色
    image = [list(map(int, input().split())) for _ in range(m)] # 图像矩阵

    old = image[sr][sc]              # 起点原颜色
    if old != color:                 # 如果颜色没变，不需要 Flood Fill
        q = deque([(sr, sc)])        # 从起点开始 BFS
        image[sr][sc] = color        # 入队时立刻染色，防止重复入队
        while q:                     # 队列中都是已经染过的点
            x, y = q.popleft()       # 取出当前像素
            for dx, dy in ((1, 0), (-1, 0), (0, 1), (0, -1)): # 四方向
                nx, ny = x + dx, y + dy # 邻居坐标
                if 0 <= nx < m and 0 <= ny < n and image[nx][ny] == old: # 合法且同色
                    image[nx][ny] = color # 入队前染色
                    q.append((nx, ny))    # 继续扩展这个邻居

    for row in image:                # 输出结果矩阵
        print(*row)
```

## F. 成绩排序

题目链接：[F:成绩排序](http://dsb12.openjudge.cn/26preexam/F/)

**题目概述**

有 $n$ 名同学，成绩互不相同。给出 $m$ 个询问回答：对 $x<y$，回答 $z=1$ 表示 $x$ 比 $y$ 高，$z=0$ 表示 $x$ 不比 $y$ 高，即 $y$ 比 $x$ 高。问有多少名同学的精确排名可以确定。


**思路**

**考点分析 / 知识点回忆**

- 考点：有向图、传递闭包、排名可确定性。
- 对应知识点：Floyd 的布尔版本可以求任意两点是否可达。
- 题型信号：关系具有传递性，问能否推断、能确定多少排名。
- 常见错误：把题目当成普通排序；只看直接入度/出度，忽略间接关系。

这不是排序题，而是有向关系推断。

建图：若 $a$ 比 $b$ 高，则加边 $a\to b$。通过传递性求闭包。

Floyd 后，$reach[i][j]$ 表示可以推出 $i$ 比 $j$ 高。

某个同学 $i$ 的排名可确定，当且仅当它和其他所有同学的相对关系都能确定：

$$
\sum_{j\ne i} [reach[i][j]\lor reach[j][i]] = n-1.
$$

$n\le 50$，使用 $O(n^3)$ Floyd 最稳。

**多种解法路线**

- 解法 1：Floyd 传递闭包。代码固定，$n\le 50$ 时最稳。时间复杂度 $O(n^3+m)$；空间复杂度 $O(n^2)$。
- 解法 2：从每个点做 DFS/BFS，统计它能到达谁；反图再做一次，统计谁能到达它。时间复杂度 $O(n(n+m))$；空间复杂度 $O(n+m)$。
- 解法 3：如果关系本身已经接近完整，也可以用集合不断传递，但不如 Floyd 清楚。朴素集合闭包时间复杂度通常仍是 $O(n^3)$ 量级；空间复杂度 $O(n^2)$。

DFS/BFS 版本的想法是：一个人的排名确定，当且仅当“他能确定比哪些人高”加上“哪些人能确定比他高”一共覆盖了其他 $n-1$ 人。


```python

# DFS/BFS 可达性版本：用正图和反图判断排名是否确定
from collections import deque        # BFS 队列

n, m = map(int, input().split())     # n 个同学，m 条比较关系
g = [[] for _ in range(n)]           # 正图：u -> v 表示 u 比 v 高
rg = [[] for _ in range(n)]          # 反图：方便找“谁比我高”

for _ in range(m):                   # 读入每条比较
    x, y, z = map(int, input().split())
    if z == 1:                       # x 比 y 高
        g[x].append(y)
        rg[y].append(x)
    else:                            # y 比 x 高
        g[y].append(x)
        rg[x].append(y)

def reachable(start, graph):         # 返回 start 在 graph 中能到达的点集合
    seen = set()                     # 已经到达的点
    q = deque([start])               # 从 start 开始 BFS
    while q:
        u = q.popleft()
        for v in graph[u]:           # 枚举 u 能推出的后继
            if v not in seen:
                seen.add(v)          # 第一次到达 v
                q.append(v)
    seen.discard(start)              # 不把自己算进相对关系
    return seen

ans = 0                              # 排名可确定的人数
for i in range(n):
    lower = reachable(i, g)          # i 能推出比哪些人高
    higher = reachable(i, rg)        # 哪些人能推出比 i 高
    if len(lower | higher) == n - 1: # 和其他所有人关系都确定
        ans += 1

print(ans)                           # 输出可确定排名的人数
```



课堂推荐 Floyd，因为这题的重点是“传递关系”，三重循环正好把传递性讲清楚。


```python

n, m = map(int, input().split())     # n 个同学，m 条比较关系
reach = [[False] * n for _ in range(n)] # reach[i][j] 表示已知/可推出 i 比 j 高

for _ in range(m):                   # 读入直接比较关系
    x, y, z = map(int, input().split())
    if z == 1:                       # x 比 y 高
        reach[x][y] = True
    else:                            # y 比 x 高
        reach[y][x] = True

for k in range(n):                   # Floyd：枚举中转点 k
    for i in range(n):               # 枚举起点 i
        if reach[i][k]:              # 只有 i 能到 k 时，k 的后继才有意义
            for j in range(n):       # 枚举终点 j
                reach[i][j] = reach[i][j] or reach[k][j] # i>k 且 k>j，则 i>j

ans = 0                              # 排名可确定的人数
for i in range(n):                   # 检查每个同学
    known = 0                        # 与 i 的相对关系已知的人数
    for j in range(n):
        if i != j and (reach[i][j] or reach[j][i]): # i 比 j 高或 j 比 i 高
            known += 1
    if known == n - 1:               # 和其他所有人都能比较
        ans += 1
print(ans)                           # 输出答案
```

**分析 / 变体讨论**

若题目问“能否确定完整排名”，则需要每个人的排名都可确定，即答案是否等于 $n$。


```python

# 变体：已知 ans 和 n，判断完整排名是否唯一
n = 4                                # 总人数
ans = 4                              # 排名可确定的人数
print("unique" if ans == n else "not unique") # 所有人排名都确定，完整排名才唯一
```

## A. 5月14日

**题目概述**

给定整数 $m,n$，求闭区间 $[m,n]$ 中所有能被 $19$ 整除的数之和。


**思路**

**考点分析 / 知识点回忆**

- 考点：基础循环、条件判断、累加器。
- 对应知识点：闭区间枚举 `range(m, n + 1)`；整除判断 `x % 19 == 0`。
- 题型信号：范围很小，要求统计/求和满足条件的数。
- 常见错误：漏掉端点；把 `< n` 写成 `<= n` 的相反错误。

直接枚举即可。范围 $0<m<n<1000$，复杂度 $O(n-m+1)$ 足够。

也可以用等差数列求和，但在上机第一题中直接循环更稳。

**多种解法路线**

- 解法 1：直接循环枚举，范围小，最不容易错。设区间长度 $D=n-m+1$，时间复杂度 $O(D)$；空间复杂度 $O(1)$。

**分析 / 变体讨论**

若范围很大，可以求第一个与最后一个 $19$ 的倍数，再用等差数列求和。

## B. 云服务

**题目概述**

有 $n$ 类任务。第 $i$ 类任务共有 $a_i$ 个，专用节点最多免费处理 $r_i$ 个，超出的任务需要交给公用云，每个费用 $c$。求最小总费用。


**思路**

**考点分析 / 知识点回忆**

- 考点：线性扫描、逐项贡献、业务题转数学表达式。
- 对应知识点：总量由独立单项贡献相加；每类任务互不影响。

每类任务互相独立。第 $i$ 类需要交给公用云的数量为：

$$
\max(0,a_i-r_i).
$$

所以总费用为：

$$
 c\sum_{i=1}^n \max(0,a_i-r_i).
$$

复杂度 $O(n)$。

**多种解法路线**

- 解法 1：逐类计算超出的任务数，累加费用。最简单，也最推荐。时间复杂度 $O(n)$；若边读边算，空间复杂度 $O(1)$，若保存数组则 $O(n)$。

## C. 索菲的血瓶

题目链接：[C:索菲的血瓶](http://xlxxsjjg.openjudge.cn/2026midexam/C/)

**题目概述**

给定小灯墙高度序列 $a_1,\dots,a_n$ 和索菲墙高度序列 $b_1,\dots,b_w$。索菲墙可以整体升高或降低任意常数。问小灯墙中有多少个长度为 $w$ 的连续区间与索菲墙相同，允许整体平移。

$n,w\le 2\times 10^5$。


**思路**

**考点分析 / 知识点回忆**

- 考点：差分序列、字符串匹配、KMP。
- 对应知识点：整体平移不改变相邻差分；KMP 可在线性时间统计模式串出现次数。
- 题型信号：允许整体升高/降低，问连续区间匹配个数。
- 常见错误：直接比较原高度；忘记处理 `w=1` 的特殊情况。

两个长度为 $w$ 的序列允许整体平移匹配，当且仅当它们的相邻差分完全相同：

$$
a_i-b_i=C \iff a_{i+1}-a_i=b_{i+1}-b_i.
$$

因此将原问题转化为：在 $a$ 的差分序列中匹配 $b$ 的差分序列。

若 $w=1$，任意单个瓶塔都可以通过平移匹配，答案为 $n$。

差分匹配可以用 KMP，复杂度 $O(n+w)$。

**多种解法路线**

- 解法 1：暴力枚举每个长度为 $w$ 的窗口。先用第一个位置算平移量，再逐项检查。时间复杂度 $O((n-w+1)w)$；空间复杂度 $O(1)$。
- 解法 2：转成差分序列后暴力匹配，省掉平移量。时间复杂度 $O(n+w+(n-w+1)(w-1))$；空间复杂度 $O(n+w)$，若只保留模式和窗口可进一步降低。
- 解法 3：差分 + KMP，最终推荐。时间复杂度 $O(n+w)$；空间复杂度 $O(w)$。
- 解法 4：差分 + 哈希，也能做到接近线性，但要处理哈希碰撞，不如 KMP 稳。期望时间复杂度 $O(n+w)$；空间复杂度 $O(n)$，滚动维护时可为 $O(1)$。

暴力写法很适合先讲“为什么要差分”：


```python

# 暴力窗口匹配：完整可运行，用来理解整体平移
n, w = map(int, input().split())     # n 是主序列长度，w 是模式长度
a = list(map(int, input().split()))  # 小灯墙高度
b = list(map(int, input().split()))  # 索菲墙高度

ans = 0                              # 匹配窗口数量
for l in range(n - w + 1):           # 枚举窗口左端点 l
    delta = a[l] - b[0]              # 假设这一段整体平移 delta
    ok = True                        # 当前窗口是否匹配
    for j in range(w):               # 逐项检查窗口内每个位置
        if a[l + j] - b[j] != delta: # 平移量不一致，说明不能整体平移匹配
            ok = False
            break
    if ok:                           # 当前窗口完全匹配
        ans += 1

print(ans)                           # 输出匹配个数
```



这段代码逻辑正确，但当 $n,w$ 同阶增长时，比较次数为 $\Theta(nw)$。例如 $n=w=2\times10^5$ 时，量级达到 $\Theta(4\times10^{10})$，优化点不是“换语言”，而是发现整体平移只影响绝对高度，不影响相邻差分。


```python

def prefix_function(p):
    pi = [0] * len(p)                # pi[i] 表示 p[:i+1] 的最长相等真前后缀长度
    for i in range(1, len(p)):       # pi[0] 一定是 0，所以从 1 开始
        j = pi[i - 1]                # 先继承上一个位置的匹配长度
        while j > 0 and p[i] != p[j]: # 当前字符失配，就不断缩短候选前缀
            j = pi[j - 1]            # 回退到更短的最长前后缀
        if p[i] == p[j]:             # 如果当前字符能接上候选前缀
            j += 1                   # 最长前后缀长度加 1
        pi[i] = j                    # 记录当前位置的前缀函数值
    return pi

n, w = map(int, input().split())     # n 是小灯墙长度，w 是索菲墙长度
a = list(map(int, input().split()))  # 小灯墙高度序列
b = list(map(int, input().split()))  # 索菲墙高度序列

if w == 1:                           # 长度为 1 时，任何单点都能通过平移匹配
    print(n)
else:
    da = [a[i + 1] - a[i] for i in range(n - 1)] # 小灯墙相邻差分
    db = [b[i + 1] - b[i] for i in range(w - 1)] # 索菲墙相邻差分
    pi = prefix_function(db)         # 对模式串差分 db 预处理 KMP 前缀函数
    j = 0                            # 当前已经匹配的 db 长度
    ans = 0                          # 匹配次数
    for x in da:                     # 依次扫描主串差分 da
        while j > 0 and x != db[j]:  # 失配时，按前缀函数回退
            j = pi[j - 1]
        if x == db[j]:               # 当前差分匹配上模式串第 j 位
            j += 1                   # 已匹配长度加 1
        if j == len(db):             # 完整匹配一个 db
            ans += 1                 # 匹配次数加 1
            j = pi[j - 1]            # 回退以寻找下一个可能的重叠匹配
    print(ans)                       # 输出可整体平移匹配的窗口个数
```

**分析 / 变体讨论**

若不允许整体升降，则直接在原序列上做模式匹配。若允许整体乘以同一比例，则需要比较相邻比值，但要注意除法精度，应改用交叉相乘。

## D. 队列pk

题目链接：[D:队列pk](http://xlxxsjjg.openjudge.cn/2026midexam/D/)

**题目概述**

给定 $1\sim n$ 的一个排列，$n$ 为偶数。每次从当前队列中取出两个相邻整数，按原顺序追加到新队列末尾。所有数取完后，希望最终新队列字典序最大。


**思路**

**考点分析 / 知识点回忆**

- 考点：贪心、双向链表、堆维护候选。
- 对应知识点：字典序最大先保证第一个位置最大；删除相邻元素后需要快速找到新的相邻关系。
- 题型信号：反复从序列中删除相邻块，并要求最终字典序最优。
- 常见错误：用列表中间删除导致 $O(n^2)$；只按原始相邻关系判断，删除后没有更新。

每一步输出的下一个数一定是当前队列中某条相邻边的左端点。因此所有“当前不是最后一个”的元素都可以作为下一输出位置的候选。

为使字典序最大，应选择当前仍有右邻居的最大元素 $u$，并将 $(u,next[u])$ 这一对取出。由于排列元素互不相同，第一关键字已经决定最优。

用双向链表维护当前队列，用最大堆维护仍可能作为左端点的元素。删除一对后，只需要把新的前驱重新加入堆。

复杂度 $O(n\log n)$。

**多种解法路线**

- 解法 1：回溯搜索所有取相邻对的顺序。正确，适合小样例验证贪心。叶子数为 $(n-1)!!$，时间复杂度至少 $O((n-1)!!)$；递归栈空间 $O(n)$，若频繁复制列表可到 $O(n^2)$。
- 解法 2：每轮在线性表中找当前最大可作为左端点的数，然后删除这一对。思路正确，但列表中间删除会移动元素。时间复杂度 $O(n^2)$；空间复杂度 $O(n)$。
- 解法 3：最大堆 + 双向链表。仍然执行同一个贪心，但删除和找候选都更快。时间复杂度 $O(n\log n)$；空间复杂度 $O(n)$。

小规模搜索可以帮助确认贪心：


```python

# 小规模回溯：枚举所有删除相邻对的顺序，用来验证贪心
n = int(input())                     # 队列长度
a = list(map(int, input().split()))  # 初始排列

def dfs(seq):                        # 返回当前序列 seq 能得到的最大字典序结果
    if not seq:                      # 没有元素了，后续结果为空
        return []
    best = None                      # 当前找到的最优结果
    for i in range(len(seq) - 1):    # 枚举删除相邻对 (i, i+1)
        rest = seq[:i] + seq[i + 2:] # 删除这一对后的剩余序列
        cur = seq[i:i + 2] + dfs(rest) # 当前先输出这一对，再接后续最优结果
        if best is None or cur > best: # Python 列表按字典序比较
            best = cur
    return best

print(*dfs(a))                       # 输出搜索得到的最优队列
```



搜索会枚举所有合法删除顺序，其叶子数为 $(n-1)!!$。观察字典序后可以发现：下一次输出的第一个数越大越好，所以每轮应取“当前有右邻居的最大元素”作为左端点。


```python

import heapq                         # 用最小堆模拟最大堆：存负数

n = int(input())                     # 队列长度
a = [0] + list(map(int, input().split())) # 1-indexed，a[i] 是第 i 个位置的值

prev = [i - 1 for i in range(n + 2)] # prev[i] 表示当前链表中 i 的左邻位置
nxt = [i + 1 for i in range(n + 2)]  # nxt[i] 表示当前链表中 i 的右邻位置
nxt[n] = 0                           # 0 表示没有右邻居
alive = [True] * (n + 1)             # alive[i] 表示位置 i 还没被删除

heap = []                            # 候选左端点堆，按 a[i] 最大优先
for i in range(1, n + 1):            # 初始时除了最后一个位置都可作左端点
    if nxt[i] != 0:
        heapq.heappush(heap, (-a[i], i)) # 存 -a[i]，让最大值最先弹出

ans = []                             # 最终输出的新队列
for _ in range(n // 2):              # 每次删除两个数，共 n/2 轮
    while True:                      # 堆里可能有过期候选，需要弹到合法为止
        _, i = heapq.heappop(heap)   # 取当前值最大的候选左端点
        j = nxt[i]                   # j 是 i 当前的右邻居
        if alive[i] and j != 0 and alive[j]: # i、j 都还活着且相邻
            break

    ans.extend([a[i], a[j]])         # 按原顺序输出这一对
    L, R = prev[i], nxt[j]           # 删除 i,j 后，L 和 R 会成为新邻居
    alive[i] = alive[j] = False      # 标记 i,j 已删除
    if L != 0:                       # 更新 L 的右邻居
        nxt[L] = R
    if R != 0:                       # 更新 R 的左邻居
        prev[R] = L
    if L != 0 and R != 0:            # L 现在有右邻居，可以成为新的候选左端点
        heapq.heappush(heap, (-a[L], L))

print(*ans)                          # 输出最优新队列
```

**分析 / 变体讨论**

若元素不互异，则堆的关键字需要比较整对 $(a_i,a_{next_i})$，因为第一关键字可能相同。


```python

# 变体：若元素可重复，候选相邻对可按 pair 比较
pairs = [(4, 2), (4, 3), (3, 9)]     # 候选相邻对
print(max(pairs))                    # Python 元组按字典序比较，先比第一项再比第二项
```

## E. 算力集群部署

题目链接：[E:算力集群部署](http://xlxxsjjg.openjudge.cn/2026midexam/E/)

**题目概述**

给定目标字符串，每次可以选择一个连续区间，把这个区间全部覆盖成同一种字符，后操作会覆盖前操作。求得到目标字符串所需的最少操作次数。$n\le 50$。


**思路**

**考点分析 / 知识点回忆**

- 考点：区间 DP、覆盖操作、后操作覆盖前操作。
- 对应知识点：区间 DP 通常按长度从小到大枚举；连续重复字符可压缩。
- 题型信号：每次操作一个连续区间，要求最少操作次数。
- 常见错误：贪心按字符段数计算，忽略后续覆盖可以合并同字符操作。

这是典型区间 DP，类似 Strange Printer。

令 $dp[l][r]$ 表示打印出子串 $s[l:r+1]$ 的最少次数。

基础转移：先单独打印 $s[l]$，再打印后面：

$$
dp[l][r]\le 1+dp[l+1][r].
$$

如果存在 $k$ 使得 $s[k]=s[l]$，那么 $s[l]$ 可以和 $s[k]$ 在同一次操作中完成：

$$
dp[l][r]\le dp[l+1][k-1]+dp[k][r].
$$

连续重复字符可先压缩，不影响答案。

**多种解法路线**

- 解法 1：按字符段数贪心。很容易想到，但会错，因为后面的覆盖可以把同字符合并。扫描时间 $O(n)$；空间 $O(1)$，但不能保证正确。
- 解法 2：记忆化搜索 `solve(l,r)`。转移和区间 DP 一样，但写成递归更接近思考过程。设压缩后长度为 $r$，时间复杂度 $O(r^3)$；空间复杂度 $O(r^2)$。
- 解法 3：自底向上的区间 DP。按长度枚举，避免递归，最终推荐。时间复杂度 $O(r^3)$；空间复杂度 $O(r^2)$。
- 解法 4：先压缩连续重复字符，再做 DP。答案不变，但状态更少。压缩时间 $O(n)$，DP 时间 $O(r^3)$，总时间 $O(n+r^3)$；空间复杂度 $O(r^2)$。

记忆化版本可以作为推导：


```python

# 记忆化搜索版区间 DP：完整可运行版本
from functools import lru_cache      # 自动缓存 solve(l, r) 的结果

s = input().strip()                  # 目标字符串
compressed = []                      # 压缩连续重复字符后的结果
for ch in s:                         # 逐字符扫描
    if not compressed or compressed[-1] != ch: # 和上一段不同才保留
        compressed.append(ch)
s = ''.join(compressed)              # 压缩后的字符串
n = len(s)                           # 压缩后长度

@lru_cache(None)                     # 记忆化所有区间答案
def solve(l, r):                     # 打印 s[l:r+1] 的最少次数
    if l > r:                        # 空区间不需要打印
        return 0
    ans = solve(l + 1, r) + 1        # 方案一：单独打印 s[l]
    for k in range(l + 1, r + 1):    # 尝试把 s[l] 和后面相同字符合并打印
        if s[k] == s[l]:             # 找到相同字符位置 k
            ans = min(ans, solve(l + 1, k - 1) + solve(k, r)) # 中间单独处理，k 和 l 合并
    return ans

print(solve(0, n - 1) if n else 0)   # 空串答案为 0，否则求整个区间
```



这题最适合讲“错误贪心 -> 递归定义 -> 区间 DP”的升级路线。


```python

s = input().strip()                  # 目标字符串
compressed = []                      # 用来存压缩后的字符
for ch in s:                         # 扫描原字符串
    if not compressed or compressed[-1] != ch: # 连续相同字符只保留一个
        compressed.append(ch)
s = ''.join(compressed)              # 压缩后的字符串

n = len(s)                           # 压缩后长度
dp = [[0] * n for _ in range(n)]     # dp[l][r] 表示打印 s[l:r+1] 的最少次数
for i in range(n):                   # 长度为 1 的区间
    dp[i][i] = 1                     # 单个字符需要打印一次

for length in range(2, n + 1):       # 按区间长度从小到大递推
    for l in range(0, n - length + 1): # 枚举左端点
        r = l + length - 1           # 右端点
        dp[l][r] = dp[l + 1][r] + 1  # 先把 s[l] 单独打印一次
        for k in range(l + 1, r + 1): # 找和 s[l] 相同的位置
            if s[k] == s[l]:         # s[l] 可以和 s[k] 合并到同一次打印
                mid = dp[l + 1][k - 1] if l + 1 <= k - 1 else 0 # 中间区间为空时为 0
                dp[l][r] = min(dp[l][r], mid + dp[k][r]) # 更新最优答案

print(dp[0][n - 1] if n else 0)      # 输出整个字符串答案
```

**分析 / 变体讨论**

若每次只能覆盖长度固定的区间，则状态需要加入长度限制；若覆盖不能重叠，则会变成区间划分问题。


```python

# 变体：压缩连续重复字符
s = "AAABBBAAC"                     # 示例字符串
compressed = []                      # 存每一段的代表字符
for ch in s:                         # 逐字符扫描
    if not compressed or compressed[-1] != ch: # 与上一段不同，才开新段
        compressed.append(ch)
print(''.join(compressed))           # AAABBBAAC -> ABAC
```

## F. 二叉树作业

题目链接：[F:二叉树作业](http://xlxxsjjg.openjudge.cn/2026midexam/F/)

**题目概述**

二叉树用字符串编码：`0` 表示叶子，`1S` 表示一个孩子，`2S1S2` 表示两个孩子。用红、绿、蓝三色染色，父子颜色不同；若有两个孩子，两个孩子颜色也不同。求合法染色中最多蓝色节点数与最少绿色节点数。

字符串长度可达 $5\times 10^5$。


**思路**

**考点分析 / 知识点回忆**

- 考点：树编码解析、树形 DP、颜色状态。
- 对应知识点：先序编码可用栈或递归解析；树形 DP 需要枚举根节点颜色与孩子颜色。
- 题型信号：树由字符串递归定义，目标是最大/最小某种颜色数量。
- 常见错误：递归深度过大；没有区分“最多蓝色”和“最少绿色”是两个不同 DP 目标。

先根据先序编码建出树。由于规模大，使用迭代建树并按编号逆序做 DP。

对每个节点 $u$ 和颜色 $c$，维护：

- $mx[u][c]$：当 $u$ 染成 $c$ 时，子树最多蓝色数；
- $mn[u][c]$：当 $u$ 染成 $c$ 时，子树最少绿色数。

叶子：

$$
mx[u][c]=[c=Blue],\quad mn[u][c]=[c=Green].
$$

一个孩子：孩子颜色不能等于父亲颜色。

两个孩子：父亲和两个孩子三者颜色必须两两不同。

**多种解法路线**

- 解法 1：递归解析树 + 递归 DP。代码直观，适合小数据，但字符串长度 $5\times10^5$ 时递归深度可能达到 $O(N)$。时间复杂度 $O(N)$；存树空间 $O(N)$，递归栈最坏 $O(N)$。
- 解法 2：栈迭代建树 + 逆序 DP。更稳，是最终写法。时间复杂度 $O(N)$；空间复杂度 $O(N)$。
- 解法 3：不显式建树，解析时直接返回子树 DP。思路漂亮，但递归深度仍是风险。时间复杂度 $O(N)$；空间复杂度 $O(H)$，其中 $H$ 是树高，最坏 $O(N)$。

递归版的核心状态很短：


```python

# 递归解析 + 递归 DP：思路直观，但深树可能递归过深
import sys
sys.setrecursionlimit(1_000_000)     # 树深时需要提高递归限制

S = input().strip()                  # 先序编码字符串，每个字符是 0/1/2

def parse(pos):                      # 从 S[pos] 开始解析一棵子树
    arity = int(S[pos])              # 当前节点孩子个数
    pos += 1                         # 消耗当前节点字符
    children = []                    # 当前节点的孩子子树
    for _ in range(arity):           # 递归解析每个孩子
        child, pos = parse(pos)
        children.append(child)
    return children, pos             # 返回子树结构和下一个未读位置

def dfs(node):                       # 返回当前子树在 3 种根颜色下的 mx/mn
    child_dp = [dfs(child) for child in node] # 先算所有孩子的 DP
    mx = [0] * 3                     # mx[c]：根染 c 时最多蓝色数
    mn = [0] * 3                     # mn[c]：根染 c 时最少绿色数

    for c in range(3):               # 枚举当前节点颜色：0红 1绿 2蓝
        base_mx = 1 if c == 2 else 0 # 当前节点是否贡献一个蓝色
        base_mn = 1 if c == 1 else 0 # 当前节点是否贡献一个绿色
        if len(child_dp) == 0:       # 叶子节点
            mx[c] = base_mx
            mn[c] = base_mn
        elif len(child_dp) == 1:     # 只有一个孩子：孩子颜色不能等于父亲
            cmx, cmn = child_dp[0]
            mx[c] = base_mx + max(cmx[d] for d in range(3) if d != c)
            mn[c] = base_mn + min(cmn[d] for d in range(3) if d != c)
        else:                        # 两个孩子：父亲和两个孩子颜色两两不同
            (amx, amn), (bmx, bmn) = child_dp
            other = [d for d in range(3) if d != c] # 两个孩子只能用剩下两种颜色
            x, y = other
            mx[c] = base_mx + max(amx[x] + bmx[y], amx[y] + bmx[x])
            mn[c] = base_mn + min(amn[x] + bmn[y], amn[y] + bmn[x])
    return mx, mn

root, _ = parse(0)                   # 解析整棵树
mx, mn = dfs(root)                   # 计算根节点 DP
print(max(mx), min(mn))              # 根可任意染色，取最多蓝色和最少绿色
```



最终代码采用迭代，是因为本题规模大；不是因为递归思路错，而是 Python 的调用栈不适合特别深的树。


```python

S = input().strip()                  # 先序编码字符串
n = len(S)                           # 每个字符对应一个节点
children = [[] for _ in range(n)]    # children[u] 存节点 u 的孩子编号
need = []                            # need[u] 表示节点 u 还缺几个孩子
stack = []                           # 栈中存还没有收满孩子的节点

for idx, ch in enumerate(S):         # 按先序顺序依次创建节点 idx
    arity = int(ch)                  # 当前节点孩子个数：0/1/2
    need.append(arity)               # 记录它需要多少孩子
    if stack:                        # 如果存在等待孩子的父节点
        parent = stack[-1]           # 栈顶就是当前节点的父亲
        children[parent].append(idx) # 当前节点挂到父亲下面
        need[parent] -= 1            # 父亲少缺一个孩子
        while stack and need[stack[-1]] == 0: # 栈顶节点孩子收满了
            stack.pop()              # 弹出，继续检查它的父亲是否也收满
    if arity > 0:                    # 当前节点自己也需要孩子
        stack.append(idx)            # 入栈等待后续节点成为它的孩子

mx = [[0] * 3 for _ in range(n)]     # mx[u][c]：u 染 c 时子树最多蓝色数
mn = [[0] * 3 for _ in range(n)]     # mn[u][c]：u 染 c 时子树最少绿色数

for u in range(n - 1, -1, -1):       # 先序编号保证孩子编号大于父亲，逆序做 DP
    if not children[u]:              # 叶子节点
        for c in range(3):
            mx[u][c] = 1 if c == 2 else 0 # 蓝色贡献给 mx
            mn[u][c] = 1 if c == 1 else 0 # 绿色贡献给 mn
    elif len(children[u]) == 1:      # 只有一个孩子
        v = children[u][0]
        for c in range(3):           # 枚举父亲颜色
            mx[u][c] = (1 if c == 2 else 0) + max(mx[v][d] for d in range(3) if d != c)
            mn[u][c] = (1 if c == 1 else 0) + min(mn[v][d] for d in range(3) if d != c)
    else:                            # 两个孩子
        a, b = children[u]
        for c in range(3):           # 枚举父亲颜色
            other = [d for d in range(3) if d != c] # 两个孩子只能用剩余两色
            x, y = other
            mx[u][c] = (1 if c == 2 else 0) + max(mx[a][x] + mx[b][y], mx[a][y] + mx[b][x])
            mn[u][c] = (1 if c == 1 else 0) + min(mn[a][x] + mn[b][y], mn[a][y] + mn[b][x])

print(max(mx[0]), min(mn[0]))        # 根节点颜色任选，取全局最优
```

**分析 / 变体讨论**

若颜色数从 $3$ 扩展到 $K$，一个节点有两个孩子时就需要枚举两个不同的子颜色，复杂度会多一个 $K^2$ 因子。


```python

# 变体：枚举父色 c 下两个孩子的不同颜色
K = 4                                # 假设颜色数扩展到 4
c = 0                                # 父节点颜色固定为 0
pairs = [(a, b) for a in range(K) for b in range(K) if a != c and b != c and a != b] # 两个孩子颜色合法组合
print(pairs)                         # 输出所有可选组合
```

## G. 数据采集

题目链接：[G:数据采集](http://xlxxsjjg.openjudge.cn/2026midexam/G/)

**题目概述**

无人机从 $(1,1)$ 出发，每一步横坐标加一。状态包含高度 $y$、垂直速度 $v$、离开基准层的连续时间。每步可选择加速度 $-1,0,1$，到达新格子后可能受到风力扰动或获得奖励。要求到达 $(m,1)$ 的最大累计奖励。

$n\le100,m\le1000,k\le10$。


**思路**

**考点分析 / 知识点回忆**

- 考点：状态 DP、滚动数组、网格状态建模。
- 对应知识点：影响未来决策的信息必须进入状态；横坐标每步加一，所以可以按列推进。
- 题型信号：题面包含位置、速度、时间限制、奖励最大化。
- 常见错误：状态只记位置，漏掉速度或离线时间；边界截断和基准层重置处理不一致。

这是状态 DP。影响未来的量不只有位置，还包括速度与离线时间。

令状态为：

$$
(x,y,v,t),
$$

其中 $t$ 表示连续离开基准层的时间。因为每一步 $x$ 必定加一，可以按列滚动 DP，只保存当前列状态。

转移：选择 $a\in\{-1,0,1\}$，先令 $v'=v+a$，再到达 $y'=y+v'$。若超过上边界，截断到 $n$；若低于 $1$，该转移非法。到达 $y'=1$ 时速度重置为 $0$，离线时间重置为 $0$。

**多种解法路线**

- 解法 1：DFS 枚举所有加速度序列。每步 3 种选择，路径数为 $3^{m-1}$。时间复杂度 $O(3^{m-1})$；递归栈空间 $O(m)$。
- 解法 2：普通 DP 记完整状态 `(x,y,v,t)`。正确，但可以继续优化空间。若速度范围记为 $V$，时间复杂度 $O(mnVk)$；空间复杂度 $O(mnVk)$。
- 解法 3：按列滚动 DP，只保留当前列的 `(y,v,t)`，最终推荐。时间复杂度 $O(mnVk)$；空间复杂度 $O(nVk)$。

暴力搜索能帮助确认状态里必须放什么：


```python

# 暴力 DFS：完整可运行，只适合很小的 m，用来理解状态
n, m, k = map(int, input().split())  # 高度上限 n，列数 m，最大离线时间 k
raw = [input().split() for _ in range(n)] # 输入从高层到低层
cell = [[None] * (m + 1) for _ in range(n + 1)] # cell[y][x] 存第 x 列高度 y 的格子
for row in range(n):                 # row=0 是最高层
    y = n - row                      # 转成题目里的高度 y
    for x in range(1, m + 1):
        token = raw[row][x - 1]      # 例如 f3 或 r10
        cell[y][x] = (token[0], int(token[1:])) # 拆成类型和数值

NEG = -10**18                        # 不可达的极小值
V_LIMIT = n + 20 * k + 5             # 给速度设一个保守边界，防止无限扩张
best = [NEG]                         # 用列表包一层，方便在 dfs 内更新

def dfs(x, y, v, t, score):          # 当前列 x、高度 y、速度 v、离线时间 t、收益 score
    if x == m:                       # 已经到最后一列
        if (y, v, t) == (1, 0, 0):   # 最后一列必须回到基准层状态
            best[0] = max(best[0], score)
        return

    for acc in (-1, 0, 1):           # 枚举本步加速度
        nv = v + acc                 # 新速度
        ny = y + nv                  # 新高度
        if ny > n:                   # 超过最高层则截断到 n
            ny = n
        if ny < 1:                   # 低于基准层非法
            continue

        if ny == 1:                  # 回到基准层：速度和离线时间清零
            dfs(x + 1, 1, 0, 0, score)
        else:
            nt = t + 1               # 离开基准层时间增加
            if nt > k:               # 超过离线限制非法
                continue
            typ, num = cell[ny][x + 1] # 查看到达的新格子
            if typ == 'f':           # 风力格：改变速度，不加收益
                nv2 = nv + num
                if abs(nv2) <= V_LIMIT:
                    dfs(x + 1, ny, nv2, nt, score)
            else:                    # 奖励格：速度不变，收益增加
                dfs(x + 1, ny, nv, nt, score + num)

dfs(1, 1, 0, 0, 0)                   # 从第 1 列、基准层、速度 0、收益 0 开始
print(best[0])                       # 输出最大收益
```



该搜索树共有 $3^{m-1}$ 条加速度序列；许多序列会到达同一个 `(x,y,v,t)`。DP 的本质就是：同一个状态只保留当前最大收益。


```python

n, m, k = map(int, input().split())  # 高度上限 n，列数 m，最大离线时间 k
raw = [input().split() for _ in range(n)] # 输入从高层到低层
cell = [[None] * (m + 1) for _ in range(n + 1)] # cell[y][x] 存格子信息
for row in range(n):
    y = n - row                      # 把输入行号转成高度
    for x in range(1, m + 1):
        token = raw[row][x - 1]
        cell[y][x] = (token[0], int(token[1:])) # (类型, 数值)

NEG = -10**18                        # 不可达状态的默认收益
V_LIMIT = n + 20 * k + 5             # 速度范围的安全限制
cur = {(1, 0, 0): 0}                 # 当前列状态：(高度, 速度, 离线时间) -> 最大收益

for x in range(1, m):                # 从第 x 列转移到第 x+1 列
    nxt = {}                         # 下一列状态表
    for (y, v, t), val in cur.items(): # 枚举当前所有可达状态
        for acc in (-1, 0, 1):       # 枚举加速度
            nv = v + acc             # 更新速度
            ny = y + nv              # 更新高度
            if ny > n:               # 超过上边界，截断
                ny = n
            if ny < 1:               # 低于基准层，非法
                continue

            if ny == 1:              # 到达基准层，状态重置
                state = (1, 0, 0)
                gain = 0
            else:
                nt = t + 1           # 离线时间增加
                if nt > k:           # 超过限制，非法
                    continue
                typ, num = cell[ny][x + 1] # 到达格子的信息
                if typ == 'f':       # 风力格：改变速度
                    nv2 = nv + num
                    if abs(nv2) > V_LIMIT:
                        continue
                    state = (ny, nv2, nt)
                    gain = 0
                else:                # 奖励格：获得收益
                    state = (ny, nv, nt)
                    gain = num

            if val + gain > nxt.get(state, NEG): # 同一状态只保留最大收益
                nxt[state] = val + gain
    cur = nxt                        # 滚动到下一列

print(cur.get((1, 0, 0), NEG))       # 最后必须回到基准层状态
```

**分析 / 变体讨论**

若没有速度和离线时间约束，问题会退化为普通网格 DP；若横坐标不固定增加，则会变成一般图上的最短路/最长路问题，需要重新考虑环。


```python

# 变体：普通从左到右网格最大收益 DP
n, m = 2, 3                          # n 行 m 列
reward = [[0, 1, 2], [0, 5, 1]]      # reward[y][x] 是到达该格收益
dp = [[-10**9] * m for _ in range(n)] # dp[y][x] 表示到达 (y,x) 的最大收益
dp[0][0] = 0                         # 起点收益为 0
for x in range(m - 1):               # 从左往右推进列
    for y in range(n):               # 枚举当前行
        for ny in range(n):          # 枚举下一列可以到达的行
            dp[ny][x + 1] = max(dp[ny][x + 1], dp[y][x] + reward[ny][x + 1])
print(max(dp[y][m - 1] for y in range(n))) # 最后一列任意行的最大收益
```

## A. 数字串的倒序输出

题目链接：[A:数字串的倒序输出](http://xlxxsjjg.openjudge.cn/2025midtest/A/)

**题目概述**

输入一串非负整数，以 `0` 结束。把 `0` 之前的数字倒序输出。

**思路 / 考点回忆**

- 考点：哨兵输入、列表/栈、反向输出。
- 不变量：遇到 `0` 之前读到的数都需要保存；`0` 是结束标记，不参与输出。
- 复杂度：若共有 $n$ 个有效数字，时间 $O(n)$，空间 $O(n)$。

```python

nums = list(map(int, input().split())) # 读入一行整数
ans = []                              # 存 0 之前的有效数字

for x in nums:                        # 从左到右扫描
    if x == 0:                        # 0 是结束标记，不参与输出
        break
    ans.append(x)                     # 保存有效数字

print(*ans[::-1])                     # 反向输出，不包含结束标记 0
```

**变体讨论**

如果数字分布在多行，可以一次读完整个输入；这里不需要保留行结构。

```python

# 变体：数字可能分布在多行
import sys

nums = list(map(int, sys.stdin.read().split())) # 读入所有 token
ans = []                              # 保存 0 之前的数字
for x in nums:
    if x == 0:                        # 遇到哨兵就停止
        break
    ans.append(x)

print(*ans[::-1])                     # 倒序输出
```

## B. 喝喝喝

题目链接：[B:喝喝喝](http://xlxxsjjg.openjudge.cn/2025midtest/B/)

**题目概述**

有 $n$ 个瓶子，每个位置为 `0/1`，一次可以选择连续区间 $[l,r]$ 把其中所有可乐喝完。问如果想一次喝完所有可乐，最少需要拿起多少个瓶子。

**思路 / 考点回忆**

- 考点：线性扫描、首尾定位、最小覆盖区间。
- 数学表达：如果第一个 `1` 在 $L$，最后一个 `1` 在 $R$，答案是 $R-L+1$；没有 `1` 时答案为 $0$。
- 复杂度：时间 $O(n)$，空间 $O(1)$。

这题的核心不是“数 1 的个数”，而是“一个连续区间必须覆盖所有 1”。

```python

n = int(input())                      # 瓶子数量
a = list(map(int, input().split()))   # 1 表示有可乐，0 表示空

first = -1                            # 第一个 1 的位置，-1 表示还没遇到
last = -1                             # 最后一个 1 的位置
for i, x in enumerate(a):             # 扫描所有瓶子
    if x == 1:
        if first == -1:               # 第一次遇到 1
            first = i
        last = i                      # 每次遇到 1 都更新 last

print(0 if first == -1 else last - first + 1) # 没有 1 则答案 0，否则覆盖 [first,last]
```

**变体讨论**

如果允许喝多次，那么答案会变成连续 `1` 段的数量；如果一次操作有代价，则可能转化为区间 DP 或贪心。

```python

# 变体：统计连续 1 段的个数
n = int(input())                      # 序列长度
a = list(map(int, input().split()))   # 0/1 序列

segments = 0                          # 连续 1 段数量
for i, x in enumerate(a):             # 枚举每个位置
    if x == 1 and (i == 0 or a[i - 1] == 0): # 当前是 1，且是某段开头
        segments += 1

print(segments)                       # 输出 1 段数量
```

## C. 不完整的纸条

题目链接：[C:不完整的纸条](http://xlxxsjjg.openjudge.cn/2025midtest/C/)

**题目概述**

给出长度为 $L$ 的字符串 $s$。它是某个周期串无限重复后截下来的一段，问最短可能周期长度。

**思路 / 考点回忆**

- 考点：KMP 前缀函数、border、最短周期。
- 周期长度 $p$ 可行，当且仅当对所有 $0\le i<L-p$，都有 $s_i=s_{i+p}$。
- KMP 的最长 border 长度为 $\pi[L-1]$，于是最短可行周期为

$$
p=L-\pi[L-1].
$$

这里不要求 $p$ 整除 $L$，因为纸条两端可能被撕掉。例如 `cabcabca` 的最短周期是 `abc`，长度为 $3$。

```python

L = int(input())                      # 字符串长度
s = input().strip()                   # 字符串

pi = [0] * L                          # pi[i] 表示 s[:i+1] 的最长相等真前后缀长度
for i in range(1, L):                 # 从第二个字符开始计算
    j = pi[i - 1]                     # 继承上一个位置的候选长度
    while j > 0 and s[i] != s[j]:     # 失配时回退
        j = pi[j - 1]
    if s[i] == s[j]:                  # 能接上当前字符
        j += 1
    pi[i] = j                         # 保存当前位置结果

print(L - pi[-1])                     # 最小循环节长度候选
```

**变体讨论**

经典“完整字符串最小循环节”通常还会判断 $L \bmod p=0$；本题不能加这个判断，因为输入只是周期串的一段子串。

```python

# 变体：完整字符串是否由某个更短串重复构成
s = input().strip()                   # 输入字符串
n = len(s)                            # 字符串长度
pi = [0] * n                          # KMP 前缀函数
for i in range(1, n):
    j = pi[i - 1]
    while j > 0 and s[i] != s[j]:     # 失配回退
        j = pi[j - 1]
    if s[i] == s[j]:                  # 匹配则扩展长度
        j += 1
    pi[i] = j

p = n - pi[-1]                        # 最短周期长度候选
print(p if n % p == 0 else n)         # 能整除才是真周期，否则周期就是 n
```

## D. 二叉树修剪

题目链接：[D:二叉树修剪](http://xlxxsjjg.openjudge.cn/2025midtest/D/)

**题目概述**

给一棵带边权的二叉树，要求保留指定数量的树枝，使保留下来的苹果数最大。

**思路 / 考点回忆**

- 考点：树形 DP、树上背包、子树合并。
- 状态：$dp[u][q]$ 表示在以 $u$ 为根的子树中保留 $q$ 条边时的最大收益。
- 合并孩子时枚举当前已用边数和孩子子树保留边数。
- 复杂度：朴素合并约 $O(NQ^2)$，空间 $O(NQ)$。

```python

import sys
sys.setrecursionlimit(10000)          # 树形 DP 需要递归

N, Q = map(int, input().split())      # N 个节点，保留 Q 条边
tree = [[] for _ in range(N + 1)]     # 邻接表
for _ in range(N - 1):                # 读入 N-1 条边
    u, v, w = map(int, input().split())
    tree[u].append((v, w))            # 无向边 u-v，权值 w
    tree[v].append((u, w))

NEG = -10**18                         # 不可达状态
dp = [[NEG] * (Q + 1) for _ in range(N + 1)] # dp[u][j]：u 子树保留 j 条边的最大苹果数
size = [0] * (N + 1)                  # size[u]：u 子树中最多可用的边数

def dfs(u, parent):                   # 以 u 为根做树形背包
    dp[u][0] = 0                      # 保留 0 条边，收益 0
    for v, w in tree[u]:              # 枚举孩子
        if v == parent:               # 不走回父亲
            continue
        dfs(v, u)                     # 先处理孩子子树
        ndp = dp[u][:]                # 合并孩子前，先复制当前状态
        for i in range(0, min(size[u], Q) + 1): # i 条边分给已合并部分
            if dp[u][i] == NEG:
                continue
            for j in range(0, min(size[v], Q - i - 1) + 1): # j 条边分给 v 子树
                ndp[i + j + 1] = max(ndp[i + j + 1], dp[u][i] + dp[v][j] + w) # 加上边 u-v
        size[u] += size[v] + 1        # 合并后，u 子树可用边数增加
        dp[u] = ndp                   # 更新 u 的 DP

dfs(1, 0)                             # 以 1 为根
print(dp[1][Q])                       # 根子树保留 Q 条边的最大收益
```

**变体讨论**

如果题目要求“最多保留 $Q$ 条边”，最后输出 `max(dp[1][:Q+1])`；如果输入已经是有根树，则可以省去 `parent` 判断。

```python

# 变体：恰好 Q 改成最多 Q
best_for_root = [0, 5, 9, 8]          # best_for_root[j] 表示保留 j 条边的收益
Q = 3                                 # 最多保留 3 条边
print(max(best_for_root[:Q + 1]))     # 在 0..Q 中取最大值
```

## E. 物流仓储选址

题目链接：[E:物流仓储选址](http://xlxxsjjg.openjudge.cn/2025midtest/E/)

**题目概述**

一条线上有 $n$ 个工厂，相邻距离已知。选择 $m$ 个工厂建立中转站，使所有工厂到最近中转站的距离和最小。

**思路 / 考点回忆**

- 考点：一维多设施选址、区间代价、中位数、分段 DP。
- 一个中转站服务连续一段工厂时，最优位置在这一段的中位数。
- 先预处理 $cost[l][r]$：用一个站服务 $l\ldots r$ 的最小距离和。
- 再做 DP：

$$
dp[k][i]=\min_{t<i}\{dp[k-1][t]+cost[t+1][i]\}.
$$

$n<500$，$O(mn^2)$ 的分段 DP 足够。

```python

n, m = map(int, input().split())      # n 个村庄，建 m 个邮局
d = list(map(int, input().split()))   # 相邻村庄距离，共 n-1 个

pos = [0] * (n + 1)                   # pos[i] 表示第 i 个村庄坐标，1-indexed
for i in range(2, n + 1):             # 根据相邻距离还原坐标
    pos[i] = pos[i - 1] + d[i - 2]

cost = [[0] * (n + 1) for _ in range(n + 1)] # cost[l][r]：一个邮局服务 l..r 的最小距离和
for l in range(1, n + 1):
    for r in range(l + 1, n + 1):
        mid = (l + r) // 2            # 一维绝对值距离，建在中位数位置最优
        cost[l][r] = cost[l][r - 1] + pos[r] - pos[mid] # 递推加入右端点 r

INF = 10**18                          # 极大值
dp = [[INF] * (n + 1) for _ in range(m + 1)] # dp[k][i]：前 i 个村庄建 k 个邮局的最小距离
dp[0][0] = 0                          # 0 个村庄建 0 个邮局，代价 0

for k in range(1, m + 1):             # 枚举邮局数量
    for i in range(1, n + 1):         # 前 i 个村庄
        for t in range(k - 1, i):     # 最后一个邮局服务 t+1..i
            dp[k][i] = min(dp[k][i], dp[k - 1][t] + cost[t + 1][i])

print(dp[m][n])                       # 输出最小总距离
```

**变体讨论**

如果所有相邻距离都为 $1$，`pos[i]=i-1`；如果 $n$ 更大，可以继续学习四边形不等式优化或分治优化 DP。

```python

# 变体：一个站服务 [l,r] 时，中位数位置最优
pos = [0, 0, 3, 4, 7]                 # 1-indexed 村庄坐标
l, r = 1, 4                           # 服务区间
mid = (l + r) // 2                    # 中位数下标
print(sum(abs(pos[i] - pos[mid]) for i in range(l, r + 1))) # 到中位数的距离和最小
```

## F. D++

题目链接：[F:D++](http://xlxxsjjg.openjudge.cn/2025midtest/F/)

**题目概述**

给出若干段伪代码，包含 `F var x y` 与 `E`。判断语法是否错误，并判断实际复杂度是否等于声称的复杂度。

**思路 / 考点回忆**

- 考点：栈模拟、变量作用域、复杂度指数维护。
- `F` 入栈，`E` 出栈；同一时刻不能重复使用同一个循环变量。
- 只有形如常数到 `n` 的有效循环，才贡献一层 $n$。
- 如果外层循环本身不执行，内层复杂度不应计入。

```python

T = int(input())                      # 测试组数

def parse_complexity(s):              # 把复杂度字符串转成 n 的指数
    if s == 'O(1)':
        return 0
    if s == 'O(n)':
        return 1
    return int(s.split('^')[1][:-1])  # O(n^w) 提取 w

def greater(x, y):                    # 判断循环下界 x 是否大于上界 y
    if x == 'n' and y != 'n':         # n > 常数
        return True
    if x != 'n' and y != 'n' and int(x) > int(y): # 常数和常数比较
        return True
    return False

for _ in range(T):
    L, comp = input().split()         # L 行代码，声称复杂度 comp
    L = int(L)
    target = parse_complexity(comp)   # 目标指数

    stack = []                        # (变量名, 是否贡献复杂度, 当前循环是否会执行)
    active_vars = set()               # 当前作用域中正在使用的循环变量
    cur_exp = 0                       # 当前有效嵌套贡献的 n 指数
    max_exp = 0                       # 过程中出现过的最大指数
    err = False                       # 是否出现语法错误

    for _ in range(L):
        parts = input().split()       # 当前语句
        if parts[0] == 'F':           # 循环开始
            _, var, x, y = parts
            if var in active_vars:    # 同一作用域内变量重复
                err = True
            active_vars.add(var)      # 新变量进入作用域

            parent_active = stack[-1][2] if stack else True # 父循环是否会执行
            loop_active = parent_active and not greater(x, y) # 当前循环是否会执行
            contributes = loop_active and x != 'n' and y == 'n' # 常数到 n 才贡献一层 n
            if contributes:
                cur_exp += 1
                max_exp = max(max_exp, cur_exp)
            stack.append((var, contributes, loop_active)) # 入栈等待匹配 E
        else:                         # 循环结束 E
            if not stack:             # 没有可匹配的 F
                err = True
            else:
                var, contributes, _ = stack.pop() # 弹出最近的 F
                active_vars.remove(var) # 变量离开作用域
                if contributes:
                    cur_exp -= 1      # 离开一个贡献 n 的循环

    if stack:                         # 还有 F 没有被 E 匹配
        err = True

    if err:
        print('ERR')                  # 语法错误优先输出 ERR
    else:
        print('Yes' if max_exp == target else 'No') # 比较实际复杂度指数
```

**变体讨论**

若只考复杂度字符串解析，先把 `O(1)`、`O(n)`、`O(n^k)` 统一转成指数 $0,1,k$，后面比较就很清楚。

```python

def parse_complexity(s):              # 把复杂度字符串转成 n 的指数
    if s == 'O(1)':                   # 常数复杂度，指数为 0
        return 0
    if s == 'O(n)':                   # 一次方
        return 1
    return int(s.split('^')[1][:-1])  # O(n^3) -> 3

for s in ['O(1)', 'O(n)', 'O(n^3)']:  # 示例测试
    print(parse_complexity(s))
```

## G. 填数争分

题目链接：[G:填数争分](http://xlxxsjjg.openjudge.cn/2025midtest/G/)

**题目概述**

给一个 9×9 数独，空格为 `0`。每个格子有权值，越靠近中心权值越高。求所有合法填法中的最高得分；无解输出 `-1`。

**思路 / 考点回忆**

- 考点：DFS 回溯、位运算候选集、最少候选优先剪枝。
- 行、列、宫分别用 bit mask 记录已经出现的数字。
- 每次选择候选数最少的空格，可以显著减少搜索树。
- 权值公式：

$$
w(r,c)=6+\min(r,c,8-r,8-c).
$$

```python

grid = [list(map(int, input().split())) for _ in range(9)] # 读入 9x9 数独
score = [[6 + min(r, c, 8 - r, 8 - c) for c in range(9)] for r in range(9)] # 靶形权值

row = [0] * 9                         # row[r] 用 bit 记录第 r 行已有数字
col = [0] * 9                         # col[c] 用 bit 记录第 c 列已有数字
box = [0] * 9                         # box[b] 用 bit 记录第 b 个 3x3 宫已有数字
empty = []                            # 空格列表
base_score = 0                        # 已填数字的初始得分
ok = True                             # 初始盘面是否合法

def box_id(r, c):                     # 计算 (r,c) 属于哪个 3x3 宫
    return (r // 3) * 3 + c // 3

for r in range(9):                    # 扫描所有格子
    for c in range(9):
        x = grid[r][c]
        if x == 0:                    # 0 表示待填
            empty.append((r, c))
        else:
            bit = 1 << x              # 用第 x 位表示数字 x
            b = box_id(r, c)
            if (row[r] & bit) or (col[c] & bit) or (box[b] & bit): # 行/列/宫重复
                ok = False
            row[r] |= bit             # 标记行已有 x
            col[c] |= bit             # 标记列已有 x
            box[b] |= bit             # 标记宫已有 x
            base_score += score[r][c] * x # 累加已填数字分数

FULL = sum(1 << x for x in range(1, 10)) # 数字 1..9 对应的完整 bit 集合
best = -1                              # 最大得分，-1 表示无解

def dfs(cur_score):                    # 当前得分为 cur_score，继续填空格
    global best
    if not empty:                      # 没有空格，得到一个完整解
        best = max(best, cur_score)
        return

    choose = -1                        # 选择候选数最少的空格，减少搜索分支
    choose_mask = 0                    # 该空格可填数字集合
    fewest = 10                        # 当前最少候选数
    for i, (r, c) in enumerate(empty): # 遍历所有空格
        mask = FULL & ~(row[r] | col[c] | box[box_id(r, c)]) # 可填数字
        cnt = mask.bit_count()         # 候选数字数量
        if cnt < fewest:
            fewest = cnt
            choose = i
            choose_mask = mask
            if cnt == 1:               # 已经最优，提前停止
                break
    if fewest == 0:                    # 某格无数可填，剪枝
        return

    r, c = empty.pop(choose)           # 取出要填的格子
    b = box_id(r, c)
    mask = choose_mask
    while mask:                        # 枚举 mask 中的每个数字
        bit = mask & -mask             # 取最低位 1
        x = bit.bit_length() - 1       # bit 转回数字 x
        row[r] |= bit                  # 放入数字 x：更新行
        col[c] |= bit                  # 更新列
        box[b] |= bit                  # 更新宫
        dfs(cur_score + score[r][c] * x) # 递归填下一个格子
        row[r] ^= bit                  # 回溯：撤销行标记
        col[c] ^= bit                  # 回溯：撤销列标记
        box[b] ^= bit                  # 回溯：撤销宫标记
        mask -= bit                    # 删除已经枚举的 bit
    empty.insert(choose, (r, c))       # 回溯：把空格放回列表

if ok:                                 # 初始盘面合法才搜索
    dfs(base_score)
print(best)                            # 输出最大得分，无解为 -1
```

**变体讨论**

如果只是普通数独，可以在找到第一个解时立即返回；本题要最大化分数，所以必须搜索所有可行填法，或加入更强的上界剪枝。

```python

# 变体：计算靶形数独权值矩阵
score = [[6 + min(r, c, 8 - r, 8 - c) for c in range(9)] for r in range(9)] # 越靠中心权值越高
for row in score:                      # 逐行输出权值矩阵
    print(*row)
```
